2. Sentiment vs Closed PnL


```python
# ============================================================
# PHASE 4 - STEP 2: SENTIMENT VS CLOSED PnL ANALYSIS
#
# Objective:
# Analyze profitability across sentiment categories.
#
# Features:
# - classification
# - Closed PnL
#
# Outputs:
# - Mean PnL by Sentiment
# - Median PnL by Sentiment
# - Total PnL by Sentiment
# - Profit/Loss Summary
# - Boxplot
# - Bar Charts
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def calculate_sentiment_pnl_statistics(dataframe):
    """
    Calculate profitability statistics grouped by sentiment.
    """
    sentiment_statistics = (
        dataframe
        .groupby("classification")["Closed PnL"]
        .agg(
            Trade_Count="count",
            Mean_PnL="mean",
            Median_PnL="median",
            Total_PnL="sum",
            Std_PnL="std",
            Min_PnL="min",
            Max_PnL="max"
        )
        .reset_index()
    )

    return sentiment_statistics


def calculate_profit_loss_by_sentiment(dataframe):
    """
    Calculate profit and loss trade counts for each sentiment.
    """
    profit_loss_summary = (
        dataframe
        .assign(
            Trade_Result=dataframe["Closed PnL"].apply(
                lambda value: (
                    "Profit"
                    if value > 0
                    else "Loss"
                    if value < 0
                    else "Breakeven"
                )
            )
        )
        .groupby(
            ["classification", "Trade_Result"]
        )
        .size()
        .unstack(fill_value=0)
        .reset_index()
    )

    return profit_loss_summary


def identify_best_and_worst_sentiment(sentiment_statistics):
    """
    Identify best and worst sentiment based on mean PnL.
    """
    best_sentiment = sentiment_statistics.loc[
        sentiment_statistics["Mean_PnL"].idxmax()
    ]

    worst_sentiment = sentiment_statistics.loc[
        sentiment_statistics["Mean_PnL"].idxmin()
    ]

    return best_sentiment, worst_sentiment


def plot_mean_pnl_by_sentiment(sentiment_statistics):
    """
    Plot average PnL by sentiment.
    """
    plt.figure(figsize=(10, 5))

    plt.bar(
        sentiment_statistics["classification"],
        sentiment_statistics["Mean_PnL"]
    )

    plt.title("Average Closed PnL by Sentiment")
    plt.xlabel("Sentiment")
    plt.ylabel("Mean Closed PnL")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


def plot_total_pnl_by_sentiment(sentiment_statistics):
    """
    Plot total PnL by sentiment.
    """
    plt.figure(figsize=(10, 5))

    plt.bar(
        sentiment_statistics["classification"],
        sentiment_statistics["Total_PnL"]
    )

    plt.title("Total Closed PnL by Sentiment")
    plt.xlabel("Sentiment")
    plt.ylabel("Total Closed PnL")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


def plot_pnl_boxplot_by_sentiment(dataframe):
    """
    Plot PnL distribution by sentiment.
    """
    sentiment_categories = sorted(
        dataframe["classification"].dropna().unique()
    )

    pnl_data = [
        dataframe.loc[
            dataframe["classification"] == sentiment,
            "Closed PnL"
        ]
        for sentiment in sentiment_categories
    ]

    plt.figure(figsize=(12, 6))

    plt.boxplot(
        pnl_data,
        labels=sentiment_categories,
        showfliers=False
    )

    plt.title("Closed PnL Distribution by Sentiment")
    plt.xlabel("Sentiment")
    plt.ylabel("Closed PnL")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

sentiment_pnl_statistics = (
    calculate_sentiment_pnl_statistics(
        phase4_df
    )
)

profit_loss_by_sentiment = (
    calculate_profit_loss_by_sentiment(
        phase4_df
    )
)

best_sentiment, worst_sentiment = (
    identify_best_and_worst_sentiment(
        sentiment_pnl_statistics
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 4 - STEP 2: SENTIMENT VS CLOSED PnL ANALYSIS")
print("=" * 70)

print("\nSENTIMENT PnL STATISTICS")
print("-" * 70)

display(sentiment_pnl_statistics)

print("\nPROFIT / LOSS SUMMARY")
print("-" * 70)

display(profit_loss_by_sentiment)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("BUSINESS INSIGHTS")
print("=" * 70)

print(
    f"Best Performing Sentiment: "
    f"{best_sentiment['classification']}"
)

print(
    f"Average PnL: "
    f"{best_sentiment['Mean_PnL']:.2f}"
)

print(
    f"Worst Performing Sentiment: "
    f"{worst_sentiment['classification']}"
)

print(
    f"Average PnL: "
    f"{worst_sentiment['Mean_PnL']:.2f}"
)

print(
    "\nObservation: Compare average and total PnL "
    "across sentiment categories to understand "
    "whether trader profitability is associated "
    "with market sentiment."
)

# ============================================================
# VISUALIZATIONS
# ============================================================

plot_mean_pnl_by_sentiment(
    sentiment_pnl_statistics
)

plot_total_pnl_by_sentiment(
    sentiment_pnl_statistics
)

plot_pnl_boxplot_by_sentiment(
    phase4_df
)
```

3. Sentiment vs Position Size


```python
# ============================================================
# PHASE 4 - STEP 3: SENTIMENT VS POSITION SIZE ANALYSIS
#
# Objective:
# Analyze trading position sizes across sentiment categories.
#
# Features:
# - classification
# - Size USD
#
# Outputs:
# - Mean Position Size
# - Median Position Size
# - Total Volume
# - Position Size Percentiles
# - Boxplot
# - Bar Charts
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def calculate_sentiment_position_statistics(dataframe):
    """
    Calculate position size statistics grouped by sentiment.
    """
    position_statistics = (
        dataframe
        .groupby("classification")["Size USD"]
        .agg(
            Trade_Count="count",
            Mean_Size_USD="mean",
            Median_Size_USD="median",
            Total_Volume_USD="sum",
            Std_Size_USD="std",
            Min_Size_USD="min",
            Max_Size_USD="max"
        )
        .reset_index()
    )

    return position_statistics


def calculate_position_percentiles_by_sentiment(dataframe):
    """
    Calculate important position size percentiles.
    """
    percentile_summary = (
        dataframe
        .groupby("classification")["Size USD"]
        .quantile([0.25, 0.50, 0.75, 0.90, 0.95, 0.99])
        .unstack()
        .reset_index()
    )

    percentile_summary.columns = [
        "classification",
        "25th Percentile",
        "50th Percentile",
        "75th Percentile",
        "90th Percentile",
        "95th Percentile",
        "99th Percentile"
    ]

    return percentile_summary


def identify_highest_and_lowest_position_sentiment(
    position_statistics
):
    """
    Identify sentiments with highest and lowest average positions.
    """
    highest_sentiment = position_statistics.loc[
        position_statistics["Mean_Size_USD"].idxmax()
    ]

    lowest_sentiment = position_statistics.loc[
        position_statistics["Mean_Size_USD"].idxmin()
    ]

    return highest_sentiment, lowest_sentiment


def plot_mean_position_size(position_statistics):
    """
    Plot average position size by sentiment.
    """
    plt.figure(figsize=(10, 5))

    plt.bar(
        position_statistics["classification"],
        position_statistics["Mean_Size_USD"]
    )

    plt.title("Average Position Size by Sentiment")
    plt.xlabel("Sentiment")
    plt.ylabel("Mean Size USD")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


def plot_total_volume(position_statistics):
    """
    Plot total traded volume by sentiment.
    """
    plt.figure(figsize=(10, 5))

    plt.bar(
        position_statistics["classification"],
        position_statistics["Total_Volume_USD"]
    )

    plt.title("Total Trading Volume by Sentiment")
    plt.xlabel("Sentiment")
    plt.ylabel("Total Volume USD")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


def plot_position_boxplot(dataframe):
    """
    Plot position size distribution by sentiment.
    """
    sentiment_categories = sorted(
        dataframe["classification"].dropna().unique()
    )

    size_data = [
        dataframe.loc[
            dataframe["classification"] == sentiment,
            "Size USD"
        ]
        for sentiment in sentiment_categories
    ]

    plt.figure(figsize=(12, 6))

    plt.boxplot(
        size_data,
        labels=sentiment_categories,
        showfliers=False
    )

    plt.title("Position Size Distribution by Sentiment")
    plt.xlabel("Sentiment")
    plt.ylabel("Size USD")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

sentiment_position_statistics = (
    calculate_sentiment_position_statistics(
        phase4_df
    )
)

position_percentiles = (
    calculate_position_percentiles_by_sentiment(
        phase4_df
    )
)

highest_sentiment, lowest_sentiment = (
    identify_highest_and_lowest_position_sentiment(
        sentiment_position_statistics
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 4 - STEP 3: SENTIMENT VS POSITION SIZE ANALYSIS")
print("=" * 70)

print("\nPOSITION SIZE STATISTICS")
print("-" * 70)

display(sentiment_position_statistics)

print("\nPOSITION SIZE PERCENTILES")
print("-" * 70)

display(position_percentiles)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("BUSINESS INSIGHTS")
print("=" * 70)

print(
    f"Highest Average Position Size: "
    f"{highest_sentiment['classification']}"
)

print(
    f"Average Position Size: "
    f"${highest_sentiment['Mean_Size_USD']:,.2f}"
)

print()

print(
    f"Lowest Average Position Size: "
    f"{lowest_sentiment['classification']}"
)

print(
    f"Average Position Size: "
    f"${lowest_sentiment['Mean_Size_USD']:,.2f}"
)

print(
    "\nObservation: Compare average position sizes "
    "across sentiment categories to understand whether "
    "traders become more aggressive during Greed or "
    "more conservative during Fear."
)

# ============================================================
# VISUALIZATIONS
# ============================================================

plot_mean_position_size(
    sentiment_position_statistics
)

plot_total_volume(
    sentiment_position_statistics
)

plot_position_boxplot(
    phase4_df
)
```

4. Sentiment vs Trading Activity


```python
# ============================================================
# PHASE 4 - STEP 4: SENTIMENT VS TRADING ACTIVITY ANALYSIS
#
# Objective:
# Analyze trader activity across sentiment categories.
#
# Features:
# - classification
# - Account
# - Merge_Date
#
# Outputs:
# - Trade Counts
# - Unique Traders
# - Average Daily Trades
# - Activity Percentage
# - Visualizations
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def preprocess_dates(dataframe):
    """
    Convert Merge_Date to datetime.
    """
    dataframe = dataframe.copy()

    dataframe["Merge_Date"] = pd.to_datetime(
        dataframe["Merge_Date"],
        errors="coerce"
    )

    return dataframe


def calculate_activity_statistics(dataframe):
    """
    Calculate trading activity metrics by sentiment.
    """
    activity_statistics = (
        dataframe
        .groupby("classification")
        .agg(
            Trade_Count=("classification", "size"),
            Unique_Traders=("Account", "nunique")
        )
        .reset_index()
    )

    activity_statistics["Activity_Percentage"] = (
        activity_statistics["Trade_Count"]
        / activity_statistics["Trade_Count"].sum()
        * 100
    ).round(2)

    return activity_statistics


def calculate_daily_activity(dataframe):
    """
    Calculate average daily trades by sentiment.
    """
    daily_activity = (
        dataframe
        .groupby(
            ["classification", "Merge_Date"]
        )
        .size()
        .reset_index(name="Daily_Trades")
    )

    average_daily_activity = (
        daily_activity
        .groupby("classification")["Daily_Trades"]
        .mean()
        .reset_index()
    )

    average_daily_activity.columns = [
        "classification",
        "Average_Daily_Trades"
    ]

    return average_daily_activity


def identify_most_active_sentiment(
    activity_statistics
):
    """
    Identify most and least active sentiment categories.
    """
    most_active = activity_statistics.loc[
        activity_statistics["Trade_Count"].idxmax()
    ]

    least_active = activity_statistics.loc[
        activity_statistics["Trade_Count"].idxmin()
    ]

    return most_active, least_active


def plot_trade_count_by_sentiment(
    activity_statistics
):
    """
    Plot trade counts.
    """
    plt.figure(figsize=(10, 5))

    plt.bar(
        activity_statistics["classification"],
        activity_statistics["Trade_Count"]
    )

    plt.title("Trade Count by Sentiment")
    plt.xlabel("Sentiment")
    plt.ylabel("Number of Trades")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


def plot_unique_traders_by_sentiment(
    activity_statistics
):
    """
    Plot unique traders by sentiment.
    """
    plt.figure(figsize=(10, 5))

    plt.bar(
        activity_statistics["classification"],
        activity_statistics["Unique_Traders"]
    )

    plt.title("Unique Traders by Sentiment")
    plt.xlabel("Sentiment")
    plt.ylabel("Number of Traders")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


def plot_average_daily_trades(
    average_daily_activity
):
    """
    Plot average daily trades.
    """
    plt.figure(figsize=(10, 5))

    plt.bar(
        average_daily_activity["classification"],
        average_daily_activity["Average_Daily_Trades"]
    )

    plt.title("Average Daily Trades by Sentiment")
    plt.xlabel("Sentiment")
    plt.ylabel("Average Daily Trades")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

phase4_df = preprocess_dates(
    phase4_df
)

activity_statistics = (
    calculate_activity_statistics(
        phase4_df
    )
)

average_daily_activity = (
    calculate_daily_activity(
        phase4_df
    )
)

most_active_sentiment, least_active_sentiment = (
    identify_most_active_sentiment(
        activity_statistics
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 4 - STEP 4: SENTIMENT VS TRADING ACTIVITY")
print("=" * 70)

print("\nTRADING ACTIVITY STATISTICS")
print("-" * 70)

display(activity_statistics)

print("\nAVERAGE DAILY TRADES")
print("-" * 70)

display(average_daily_activity)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("BUSINESS INSIGHTS")
print("=" * 70)

print(
    f"Most Active Sentiment: "
    f"{most_active_sentiment['classification']}"
)

print(
    f"Trade Count: "
    f"{most_active_sentiment['Trade_Count']:,}"
)

print()

print(
    f"Least Active Sentiment: "
    f"{least_active_sentiment['classification']}"
)

print(
    f"Trade Count: "
    f"{least_active_sentiment['Trade_Count']:,}"
)

print()

print(
    f"Activity Share of Most Active Sentiment: "
    f"{most_active_sentiment['Activity_Percentage']:.2f}%"
)

print(
    "\nObservation: Higher activity may indicate "
    "strong trader participation during specific "
    "market sentiment conditions."
)

# ============================================================
# VISUALIZATIONS
# ============================================================

plot_trade_count_by_sentiment(
    activity_statistics
)

plot_unique_traders_by_sentiment(
    activity_statistics
)

plot_average_daily_trades(
    average_daily_activity
)
```

5. Sentiment vs Win Rate


```python
# ============================================================
# PHASE 4 - STEP 5: SENTIMENT VS WIN RATE ANALYSIS
#
# Objective:
# Analyze trader success rates across sentiment categories.
#
# Features:
# - classification
# - Closed PnL
#
# Outputs:
# - Win Rate
# - Loss Rate
# - Breakeven Rate
# - Win/Loss Counts
# - Visualizations
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def create_trade_outcome_column(dataframe):
    """
    Create trade outcome labels.
    """
    dataframe = dataframe.copy()

    dataframe["Trade_Outcome"] = dataframe["Closed PnL"].apply(
        lambda value: (
            "Win"
            if value > 0
            else "Loss"
            if value < 0
            else "Breakeven"
        )
    )

    return dataframe


def calculate_win_rate_statistics(dataframe):
    """
    Calculate outcome counts and percentages.
    """
    outcome_counts = (
        dataframe
        .groupby(
            ["classification", "Trade_Outcome"]
        )
        .size()
        .unstack(fill_value=0)
    )

    outcome_percentages = (
        outcome_counts.div(
            outcome_counts.sum(axis=1),
            axis=0
        ) * 100
    ).round(2)

    outcome_counts = outcome_counts.reset_index()
    outcome_percentages = outcome_percentages.reset_index()

    return outcome_counts, outcome_percentages


def create_win_rate_summary(outcome_percentages):
    """
    Create summary table containing
    Win, Loss and Breakeven rates.
    """
    summary = outcome_percentages.copy()

    summary = summary.rename(
        columns={
            "Win": "Win Rate (%)",
            "Loss": "Loss Rate (%)",
            "Breakeven": "Breakeven Rate (%)"
        }
    )

    return summary


def identify_best_and_worst_win_rate(
    win_rate_summary
):
    """
    Identify highest and lowest win rates.
    """
    best_sentiment = win_rate_summary.loc[
        win_rate_summary["Win Rate (%)"].idxmax()
    ]

    worst_sentiment = win_rate_summary.loc[
        win_rate_summary["Win Rate (%)"].idxmin()
    ]

    return best_sentiment, worst_sentiment


def plot_win_rate(win_rate_summary):
    """
    Plot win rates.
    """
    plt.figure(figsize=(10, 5))

    plt.bar(
        win_rate_summary["classification"],
        win_rate_summary["Win Rate (%)"]
    )

    plt.title("Win Rate by Sentiment")
    plt.xlabel("Sentiment")
    plt.ylabel("Win Rate (%)")

    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()


def plot_stacked_outcomes(outcome_percentages):
    """
    Plot stacked outcome percentages.
    """
    plot_df = outcome_percentages.set_index(
        "classification"
    )

    plot_df[
        ["Win", "Loss", "Breakeven"]
    ].plot(
        kind="bar",
        stacked=True,
        figsize=(10, 6)
    )

    plt.title(
        "Trade Outcome Distribution by Sentiment"
    )
    plt.xlabel("Sentiment")
    plt.ylabel("Percentage")

    plt.legend(
        title="Outcome"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

phase4_df = create_trade_outcome_column(
    phase4_df
)

outcome_counts, outcome_percentages = (
    calculate_win_rate_statistics(
        phase4_df
    )
)

win_rate_summary = (
    create_win_rate_summary(
        outcome_percentages
    )
)

best_sentiment, worst_sentiment = (
    identify_best_and_worst_win_rate(
        win_rate_summary
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 4 - STEP 5: SENTIMENT VS WIN RATE ANALYSIS")
print("=" * 70)

print("\nWIN / LOSS COUNTS")
print("-" * 70)

display(outcome_counts)

print("\nWIN RATE SUMMARY")
print("-" * 70)

display(win_rate_summary)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("BUSINESS INSIGHTS")
print("=" * 70)

print(
    f"Highest Win Rate Sentiment: "
    f"{best_sentiment['classification']}"
)

print(
    f"Win Rate: "
    f"{best_sentiment['Win Rate (%)']:.2f}%"
)

print()

print(
    f"Lowest Win Rate Sentiment: "
    f"{worst_sentiment['classification']}"
)

print(
    f"Win Rate: "
    f"{worst_sentiment['Win Rate (%)']:.2f}%"
)

print()

print(
    "Observation: Comparing win rates across "
    "sentiment categories helps determine "
    "whether market psychology influences "
    "trading success."
)

# ============================================================
# VISUALIZATIONS
# ============================================================

plot_win_rate(
    win_rate_summary
)

plot_stacked_outcomes(
    outcome_percentages
)
```

6. SENTIMENT VS TRADE DIRECTION ANALYSIS

```python
# ============================================================
# PHASE 4 - STEP 6: SENTIMENT VS TRADE DIRECTION ANALYSIS
#
# Objective:
# Analyze trading direction across sentiment categories.
#
# Features:
# - classification
# - Direction
#
# Outputs:
# - Direction Counts
# - Direction Percentages
# - Stacked Bar Chart
# - Heatmap
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def calculate_direction_counts(dataframe):
    """
    Calculate direction counts by sentiment.
    """
    direction_counts = (
        pd.crosstab(
            dataframe["classification"],
            dataframe["Direction"]
        )
    )

    return direction_counts


def calculate_direction_percentages(direction_counts):
    """
    Calculate direction percentages by sentiment.
    """
    direction_percentages = (
        direction_counts.div(
            direction_counts.sum(axis=1),
            axis=0
        ) * 100
    ).round(2)

    return direction_percentages


def get_top_direction_per_sentiment(
    direction_counts
):
    """
    Identify dominant direction for each sentiment.
    """
    top_directions = (
        direction_counts.idxmax(axis=1)
        .reset_index()
    )

    top_directions.columns = [
        "Sentiment",
        "Dominant Direction"
    ]

    return top_directions


def plot_direction_distribution(
    direction_percentages
):
    """
    Plot stacked bar chart.
    """
    direction_percentages.plot(
        kind="bar",
        stacked=True,
        figsize=(14, 7)
    )

    plt.title(
        "Trade Direction Distribution by Sentiment"
    )

    plt.xlabel("Sentiment")
    plt.ylabel("Percentage")

    plt.legend(
        title="Direction",
        bbox_to_anchor=(1.05, 1),
        loc="upper left"
    )

    plt.tight_layout()
    plt.show()


def plot_direction_heatmap(
    direction_percentages
):
    """
    Create heatmap using matplotlib.
    """
    plt.figure(figsize=(14, 6))

    plt.imshow(
        direction_percentages,
        aspect="auto"
    )

    plt.colorbar(
        label="Percentage"
    )

    plt.xticks(
        range(len(direction_percentages.columns)),
        direction_percentages.columns,
        rotation=90
    )

    plt.yticks(
        range(len(direction_percentages.index)),
        direction_percentages.index
    )

    plt.title(
        "Direction Percentage Heatmap by Sentiment"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

direction_counts = (
    calculate_direction_counts(
        phase4_df
    )
)

direction_percentages = (
    calculate_direction_percentages(
        direction_counts
    )
)

dominant_directions = (
    get_top_direction_per_sentiment(
        direction_counts
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 4 - STEP 6: SENTIMENT VS TRADE DIRECTION")
print("=" * 70)

print("\nDIRECTION COUNTS")
print("-" * 70)

display(direction_counts)

print("\nDIRECTION PERCENTAGES")
print("-" * 70)

display(direction_percentages)

print("\nDOMINANT DIRECTION BY SENTIMENT")
print("-" * 70)

display(dominant_directions)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("BUSINESS INSIGHTS")
print("=" * 70)

for _, row in dominant_directions.iterrows():
    print(
        f"{row['Sentiment']} → "
        f"{row['Dominant Direction']}"
    )

print(
    "\nObservation: If dominant directions differ "
    "across sentiment categories, trader behavior "
    "may be influenced by market psychology."
)

# ============================================================
# VISUALIZATIONS
# ============================================================

plot_direction_distribution(
    direction_percentages
)

plot_direction_heatmap(
    direction_percentages
)
```

7. COIN PERFORMANCE VS SENTIMENT ANALYSIS

```python
# ============================================================
# PHASE 4 - STEP 7: COIN PERFORMANCE VS SENTIMENT
#
# Objective:
# Analyze coin profitability under different
# market sentiment conditions.
#
# Features:
# - classification
# - Coin
# - Closed PnL
#
# Outputs:
# - Coin Profitability Tables
# - Top Coins
# - Worst Coins
# - Heatmap
# - Bar Charts
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def calculate_coin_sentiment_performance(dataframe):
    """
    Calculate profitability statistics for each coin
    within each sentiment category.
    """
    performance_summary = (
        dataframe
        .groupby(
            ["classification", "Coin"]
        )["Closed PnL"]
        .agg(
            Trade_Count="count",
            Mean_PnL="mean",
            Median_PnL="median",
            Total_PnL="sum"
        )
        .reset_index()
    )

    return performance_summary


def get_top_coins_by_sentiment(
    performance_summary,
    top_n=10
):
    """
    Get top performing coins for each sentiment.
    """
    top_coin_results = {}

    sentiments = (
        performance_summary["classification"]
        .unique()
    )

    for sentiment in sentiments:

        sentiment_df = (
            performance_summary[
                performance_summary["classification"]
                == sentiment
            ]
            .sort_values(
                by="Mean_PnL",
                ascending=False
            )
            .head(top_n)
        )

        top_coin_results[sentiment] = sentiment_df

    return top_coin_results


def get_worst_coins_by_sentiment(
    performance_summary,
    top_n=10
):
    """
    Get worst performing coins for each sentiment.
    """
    worst_coin_results = {}

    sentiments = (
        performance_summary["classification"]
        .unique()
    )

    for sentiment in sentiments:

        sentiment_df = (
            performance_summary[
                performance_summary["classification"]
                == sentiment
            ]
            .sort_values(
                by="Mean_PnL",
                ascending=True
            )
            .head(top_n)
        )

        worst_coin_results[sentiment] = sentiment_df

    return worst_coin_results


def create_coin_sentiment_heatmap_data(
    performance_summary
):
    """
    Create coin-sentiment matrix using Mean PnL.
    """
    heatmap_data = (
        performance_summary
        .pivot_table(
            index="Coin",
            columns="classification",
            values="Mean_PnL",
            fill_value=0
        )
    )

    return heatmap_data


def plot_top_coin_performance(
    top_coin_dataframe,
    sentiment
):
    """
    Plot top coin profitability.
    """
    plt.figure(figsize=(12, 6))

    plt.bar(
        top_coin_dataframe["Coin"],
        top_coin_dataframe["Mean_PnL"]
    )

    plt.title(
        f"Top Coin Performance During {sentiment}"
    )

    plt.xlabel("Coin")
    plt.ylabel("Average Closed PnL")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


def plot_coin_sentiment_heatmap(
    heatmap_data,
    top_coins=20
):
    """
    Plot profitability heatmap.
    """
    heatmap_subset = (
        heatmap_data
        .abs()
        .max(axis=1)
        .sort_values(ascending=False)
        .head(top_coins)
        .index
    )

    heatmap_plot_data = (
        heatmap_data
        .loc[heatmap_subset]
    )

    plt.figure(figsize=(12, 8))

    plt.imshow(
        heatmap_plot_data,
        aspect="auto"
    )

    plt.colorbar(
        label="Average Closed PnL"
    )

    plt.xticks(
        range(len(heatmap_plot_data.columns)),
        heatmap_plot_data.columns,
        rotation=45
    )

    plt.yticks(
        range(len(heatmap_plot_data.index)),
        heatmap_plot_data.index
    )

    plt.title(
        "Coin Profitability Across Sentiments"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

coin_performance_summary = (
    calculate_coin_sentiment_performance(
        phase4_df
    )
)

top_coin_results = (
    get_top_coins_by_sentiment(
        coin_performance_summary
    )
)

worst_coin_results = (
    get_worst_coins_by_sentiment(
        coin_performance_summary
    )
)

heatmap_data = (
    create_coin_sentiment_heatmap_data(
        coin_performance_summary
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 4 - STEP 7: COIN PERFORMANCE VS SENTIMENT")
print("=" * 70)

print("\nCOIN PERFORMANCE SUMMARY")
print("-" * 70)

display(
    coin_performance_summary.head(20)
)

for sentiment, dataframe in top_coin_results.items():

    print("\n" + "=" * 70)
    print(f"TOP 10 COINS DURING {sentiment}")
    print("=" * 70)

    display(
        dataframe[
            [
                "Coin",
                "Trade_Count",
                "Mean_PnL",
                "Total_PnL"
            ]
        ]
    )

for sentiment, dataframe in worst_coin_results.items():

    print("\n" + "=" * 70)
    print(f"WORST 10 COINS DURING {sentiment}")
    print("=" * 70)

    display(
        dataframe[
            [
                "Coin",
                "Trade_Count",
                "Mean_PnL",
                "Total_PnL"
            ]
        ]
    )

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("BUSINESS INSIGHTS")
print("=" * 70)

best_coin = (
    coin_performance_summary
    .sort_values(
        by="Mean_PnL",
        ascending=False
    )
    .iloc[0]
)

worst_coin = (
    coin_performance_summary
    .sort_values(
        by="Mean_PnL",
        ascending=True
    )
    .iloc[0]
)

print(
    f"Best Coin-Sentiment Combination:"
)

print(
    f"{best_coin['Coin']} during "
    f"{best_coin['classification']}"
)

print(
    f"Average PnL: "
    f"{best_coin['Mean_PnL']:.2f}"
)

print()

print(
    f"Worst Coin-Sentiment Combination:"
)

print(
    f"{worst_coin['Coin']} during "
    f"{worst_coin['classification']}"
)

print(
    f"Average PnL: "
    f"{worst_coin['Mean_PnL']:.2f}"
)

# ============================================================
# VISUALIZATIONS
# ============================================================

for sentiment, dataframe in top_coin_results.items():

    plot_top_coin_performance(
        dataframe,
        sentiment
    )

plot_coin_sentiment_heatmap(
    heatmap_data
)
```

8. TRADER PERFORMANCE VS SENTIMENT ANALYSIS

```python
# ============================================================
# PHASE 4 - STEP 8: TRADER PERFORMANCE VS SENTIMENT
#
# Objective:
# Analyze trader profitability across sentiment categories.
#
# Features:
# - classification
# - Account
# - Closed PnL
#
# Outputs:
# - Trader Rankings
# - Top Traders
# - Worst Traders
# - Trader Performance Matrix
# - Visualizations
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def calculate_trader_sentiment_performance(dataframe):
    """
    Calculate trader profitability metrics by sentiment.
    """
    trader_summary = (
        dataframe
        .groupby(
            ["classification", "Account"]
        )["Closed PnL"]
        .agg(
            Trade_Count="count",
            Mean_PnL="mean",
            Median_PnL="median",
            Total_PnL="sum"
        )
        .reset_index()
    )

    return trader_summary


def get_top_traders_by_sentiment(
    trader_summary,
    top_n=10
):
    """
    Get top performing traders for each sentiment.
    """
    top_trader_results = {}

    sentiments = (
        trader_summary["classification"]
        .unique()
    )

    for sentiment in sentiments:

        sentiment_df = (
            trader_summary[
                trader_summary["classification"]
                == sentiment
            ]
            .sort_values(
                by="Mean_PnL",
                ascending=False
            )
            .head(top_n)
        )

        top_trader_results[sentiment] = sentiment_df

    return top_trader_results


def get_worst_traders_by_sentiment(
    trader_summary,
    top_n=10
):
    """
    Get worst performing traders for each sentiment.
    """
    worst_trader_results = {}

    sentiments = (
        trader_summary["classification"]
        .unique()
    )

    for sentiment in sentiments:

        sentiment_df = (
            trader_summary[
                trader_summary["classification"]
                == sentiment
            ]
            .sort_values(
                by="Mean_PnL",
                ascending=True
            )
            .head(top_n)
        )

        worst_trader_results[sentiment] = sentiment_df

    return worst_trader_results


def create_trader_performance_matrix(
    trader_summary
):
    """
    Create trader-sentiment profitability matrix.
    """
    performance_matrix = (
        trader_summary
        .pivot_table(
            index="Account",
            columns="classification",
            values="Mean_PnL",
            fill_value=0
        )
    )

    return performance_matrix


def calculate_overall_trader_ranking(
    dataframe
):
    """
    Calculate overall trader ranking.
    """
    ranking = (
        dataframe
        .groupby("Account")["Closed PnL"]
        .agg(
            Total_Trades="count",
            Mean_PnL="mean",
            Total_PnL="sum"
        )
        .reset_index()
        .sort_values(
            by="Mean_PnL",
            ascending=False
        )
    )

    return ranking


def plot_top_traders(
    trader_dataframe,
    sentiment
):
    """
    Plot top traders for a sentiment.
    """
    plt.figure(figsize=(12, 6))

    trader_labels = [
        trader[-8:]
        for trader in trader_dataframe["Account"]
    ]

    plt.bar(
        trader_labels,
        trader_dataframe["Mean_PnL"]
    )

    plt.title(
        f"Top Traders During {sentiment}"
    )

    plt.xlabel("Trader")
    plt.ylabel("Average Closed PnL")

    plt.xticks(rotation=45)

    plt.tight_layout()
    plt.show()


def plot_trader_heatmap(
    performance_matrix,
    top_traders=15
):
    """
    Plot trader sentiment heatmap.
    """
    selected_traders = (
        performance_matrix
        .abs()
        .max(axis=1)
        .sort_values(ascending=False)
        .head(top_traders)
        .index
    )

    heatmap_data = (
        performance_matrix
        .loc[selected_traders]
    )

    plt.figure(figsize=(12, 8))

    plt.imshow(
        heatmap_data,
        aspect="auto"
    )

    plt.colorbar(
        label="Average Closed PnL"
    )

    plt.xticks(
        range(len(heatmap_data.columns)),
        heatmap_data.columns,
        rotation=45
    )

    trader_labels = [
        trader[-8:]
        for trader in heatmap_data.index
    ]

    plt.yticks(
        range(len(trader_labels)),
        trader_labels
    )

    plt.title(
        "Trader Performance Across Sentiments"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

trader_summary = (
    calculate_trader_sentiment_performance(
        phase4_df
    )
)

top_traders = (
    get_top_traders_by_sentiment(
        trader_summary
    )
)

worst_traders = (
    get_worst_traders_by_sentiment(
        trader_summary
    )
)

performance_matrix = (
    create_trader_performance_matrix(
        trader_summary
    )
)

overall_ranking = (
    calculate_overall_trader_ranking(
        phase4_df
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 4 - STEP 8: TRADER PERFORMANCE VS SENTIMENT")
print("=" * 70)

print("\nOVERALL TRADER RANKING")
print("-" * 70)

display(
    overall_ranking.head(20)
)

for sentiment, dataframe in top_traders.items():

    print("\n" + "=" * 70)
    print(f"TOP TRADERS DURING {sentiment}")
    print("=" * 70)

    display(
        dataframe[
            [
                "Account",
                "Trade_Count",
                "Mean_PnL",
                "Total_PnL"
            ]
        ]
    )

for sentiment, dataframe in worst_traders.items():

    print("\n" + "=" * 70)
    print(f"WORST TRADERS DURING {sentiment}")
    print("=" * 70)

    display(
        dataframe[
            [
                "Account",
                "Trade_Count",
                "Mean_PnL",
                "Total_PnL"
            ]
        ]
    )

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("BUSINESS INSIGHTS")
print("=" * 70)

best_trader = overall_ranking.iloc[0]
worst_trader = overall_ranking.iloc[-1]

print(
    f"Best Overall Trader:"
)

print(
    f"{best_trader['Account']}"
)

print(
    f"Average PnL: "
    f"{best_trader['Mean_PnL']:.2f}"
)

print()

print(
    f"Worst Overall Trader:"
)

print(
    f"{worst_trader['Account']}"
)

print(
    f"Average PnL: "
    f"{worst_trader['Mean_PnL']:.2f}"
)

print()

print(
    "Observation: Traders who remain profitable "
    "across all sentiment categories may represent "
    "robust trading strategies that are less dependent "
    "on market psychology."
)

# ============================================================
# VISUALIZATIONS
# ============================================================

for sentiment, dataframe in top_traders.items():

    plot_top_traders(
        dataframe,
        sentiment
    )

plot_trader_heatmap(
    performance_matrix
)
```

9B. ANOVA TEST (PROFITABILITY VS SENTIMENT)

```python
# ============================================================
# PHASE 4 - STEP 9B:
# ANOVA TEST
# PROFITABILITY VS SENTIMENT
# ============================================================

from scipy.stats import f_oneway


def perform_anova_profitability(dataframe):
    """
    ANOVA test for Closed PnL.
    """

    groups = [
        group["Closed PnL"]
        for _, group in dataframe.groupby(
            "classification"
        )
    ]

    statistic, p_value = f_oneway(*groups)

    return statistic, p_value


def get_decision(p_value):

    if p_value < 0.05:
        return "Reject H0"

    return "Fail to Reject H0"


# MAIN PROGRAM

anova_statistic, anova_pvalue = (
    perform_anova_profitability(
        phase4_df
    )
)

print("=" * 70)
print("ANOVA TEST")
print("=" * 70)

print("Statistic :", anova_statistic)
print("P-Value   :", anova_pvalue)

print(
    "Decision  :",
    get_decision(anova_pvalue)
)
```

9C. KRUSKAL-WALLIS TEST

```python
# ============================================================
# PHASE 4 - STEP 9C:
# KRUSKAL-WALLIS TEST
# ============================================================

from scipy.stats import kruskal


def perform_kruskal_profitability(dataframe):
    """
    Closed PnL vs Sentiment
    """

    groups = [
        group["Closed PnL"]
        for _, group in dataframe.groupby(
            "classification"
        )
    ]

    statistic, p_value = kruskal(*groups)

    return statistic, p_value


def perform_kruskal_position_size(dataframe):
    """
    Size USD vs Sentiment
    """

    groups = [
        group["Size USD"]
        for _, group in dataframe.groupby(
            "classification"
        )
    ]

    statistic, p_value = kruskal(*groups)

    return statistic, p_value


def get_decision(p_value):

    if p_value < 0.05:
        return "Reject H0"

    return "Fail to Reject H0"


# MAIN PROGRAM

pnl_statistic, pnl_pvalue = (
    perform_kruskal_profitability(
        phase4_df
    )
)

size_statistic, size_pvalue = (
    perform_kruskal_position_size(
        phase4_df
    )
)

print("=" * 70)
print("KRUSKAL-WALLIS TEST")
print("=" * 70)

print("\nPROFITABILITY VS SENTIMENT")
print("-" * 70)

print("Statistic :", pnl_statistic)
print("P-Value   :", pnl_pvalue)
print(
    "Decision  :",
    get_decision(pnl_pvalue)
)

print("\nPOSITION SIZE VS SENTIMENT")
print("-" * 70)

print("Statistic :", size_statistic)
print("P-Value   :", size_pvalue)
print(
    "Decision  :",
    get_decision(size_pvalue)
)
```

9D. CHI-SQUARE TEST (DIRECTION VS SENTIMENT)

```python
# ============================================================
# PHASE 4 - STEP 9D:
# CHI-SQUARE TEST
# DIRECTION VS SENTIMENT
# ============================================================

import pandas as pd

from scipy.stats import chi2_contingency


def create_contingency_table(dataframe):
    """
    Create contingency table.
    """

    contingency_table = pd.crosstab(
        dataframe["classification"],
        dataframe["Direction"]
    )

    return contingency_table


def perform_chi_square_test(
    contingency_table
):
    """
    Chi-Square test.
    """

    (
        statistic,
        p_value,
        degrees_of_freedom,
        expected
    ) = chi2_contingency(
        contingency_table
    )

    return (
        statistic,
        p_value,
        degrees_of_freedom
    )


def get_decision(p_value):

    if p_value < 0.05:
        return "Reject H0"

    return "Fail to Reject H0"


# MAIN PROGRAM

contingency_table = (
    create_contingency_table(
        phase4_df
    )
)

(
    chi_statistic,
    chi_pvalue,
    chi_dof
) = perform_chi_square_test(
    contingency_table
)

print("=" * 70)
print("CHI-SQUARE TEST")
print("=" * 70)

display(contingency_table)

print("\nStatistic :", chi_statistic)
print("P-Value   :", chi_pvalue)
print("DOF       :", chi_dof)

print(
    "Decision  :",
    get_decision(chi_pvalue)
)
```

9E. EFFECT SIZE ANALYSIS

```python
# ============================================================
# PHASE 4 - STEP 9E:
# EFFECT SIZE ANALYSIS
#
# Objective:
# Measure practical significance of sentiment effects.
#
# Metrics:
# - Eta Squared (ANOVA)
# - Epsilon Squared (Kruskal-Wallis)
# - Cramer's V (Chi-Square)
# ============================================================

import pandas as pd
import numpy as np

from scipy.stats import (
    f_oneway,
    kruskal,
    chi2_contingency
)


def calculate_eta_squared(dataframe):
    """
    Calculate Eta Squared (η²)
    for ANOVA: Closed PnL vs Sentiment.
    """

    grand_mean = dataframe["Closed PnL"].mean()

    grouped_data = dataframe.groupby(
        "classification"
    )["Closed PnL"]

    ss_between = sum(
        len(group) * (
            group.mean() - grand_mean
        ) ** 2
        for _, group in grouped_data
    )

    ss_total = sum(
        (
            dataframe["Closed PnL"]
            - grand_mean
        ) ** 2
    )

    eta_squared = ss_between / ss_total

    return eta_squared


def calculate_epsilon_squared_pnl(dataframe):
    """
    Epsilon Squared for
    Kruskal-Wallis:
    Profitability vs Sentiment.
    """

    groups = [
        group["Closed PnL"]
        for _, group in dataframe.groupby(
            "classification"
        )
    ]

    h_statistic, _ = kruskal(*groups)

    number_of_groups = len(groups)

    sample_size = len(dataframe)

    epsilon_squared = (
        h_statistic
        - number_of_groups
        + 1
    ) / (
        sample_size
        - number_of_groups
    )

    return epsilon_squared


def calculate_epsilon_squared_size(dataframe):
    """
    Epsilon Squared for
    Kruskal-Wallis:
    Position Size vs Sentiment.
    """

    groups = [
        group["Size USD"]
        for _, group in dataframe.groupby(
            "classification"
        )
    ]

    h_statistic, _ = kruskal(*groups)

    number_of_groups = len(groups)

    sample_size = len(dataframe)

    epsilon_squared = (
        h_statistic
        - number_of_groups
        + 1
    ) / (
        sample_size
        - number_of_groups
    )

    return epsilon_squared


def calculate_cramers_v(dataframe):
    """
    Calculate Cramer's V
    for Direction vs Sentiment.
    """

    contingency_table = pd.crosstab(
        dataframe["classification"],
        dataframe["Direction"]
    )

    chi_square_statistic, _, _, _ = (
        chi2_contingency(
            contingency_table
        )
    )

    sample_size = contingency_table.sum().sum()

    minimum_dimension = min(
        contingency_table.shape[0] - 1,
        contingency_table.shape[1] - 1
    )

    cramers_v = np.sqrt(
        chi_square_statistic /
        (
            sample_size *
            minimum_dimension
        )
    )

    return cramers_v


def interpret_effect_size(value):
    """
    Interpret effect size.
    """

    if value < 0.01:
        return "Negligible"

    if value < 0.06:
        return "Small"

    if value < 0.14:
        return "Medium"

    return "Large"


# ============================================================
# MAIN PROGRAM
# ============================================================

eta_squared = calculate_eta_squared(
    phase4_df
)

epsilon_squared_pnl = (
    calculate_epsilon_squared_pnl(
        phase4_df
    )
)

epsilon_squared_size = (
    calculate_epsilon_squared_size(
        phase4_df
    )
)

cramers_v = calculate_cramers_v(
    phase4_df
)

effect_size_summary = pd.DataFrame({
    "Analysis": [
        "ANOVA (PnL vs Sentiment)",
        "Kruskal-Wallis (PnL vs Sentiment)",
        "Kruskal-Wallis (Size vs Sentiment)",
        "Chi-Square (Direction vs Sentiment)"
    ],
    "Effect Size": [
        eta_squared,
        epsilon_squared_pnl,
        epsilon_squared_size,
        cramers_v
    ],
    "Interpretation": [
        interpret_effect_size(
            eta_squared
        ),
        interpret_effect_size(
            epsilon_squared_pnl
        ),
        interpret_effect_size(
            epsilon_squared_size
        ),
        interpret_effect_size(
            cramers_v
        )
    ]
})

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 4 - STEP 9E")
print("EFFECT SIZE ANALYSIS")
print("=" * 70)

display(effect_size_summary)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("INTERPRETATION")
print("=" * 70)

for _, row in effect_size_summary.iterrows():

    print(
        f"{row['Analysis']}"
    )

    print(
        f"Effect Size : "
        f"{row['Effect Size']:.6f}"
    )

    print(
        f"Magnitude   : "
        f"{row['Interpretation']}"
    )

    print("-" * 50)
```

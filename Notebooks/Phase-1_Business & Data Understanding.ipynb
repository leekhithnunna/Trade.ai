```python
# ============================================================
# DATASET 1: FEAR & GREED INDEX UNDERSTANDING
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def load_fear_greed_dataset(file_path):
    """
    Load Fear & Greed dataset.
    """
    dataframe = pd.read_csv(file_path)

    return dataframe


def preprocess_fear_greed_dataset(dataframe):
    """
    Convert date column to datetime.
    """
    dataframe = dataframe.copy()

    dataframe["date"] = pd.to_datetime(dataframe["date"])

    return dataframe


def get_fear_greed_summary(dataframe):
    """
    Generate dataset summary.
    """
    summary = {
        "Rows": dataframe.shape[0],
        "Columns": dataframe.shape[1],
        "Column Names": list(dataframe.columns),
        "Date Range Start": dataframe["date"].min(),
        "Date Range End": dataframe["date"].max(),
        "Sentiment Categories": sorted(
            dataframe["classification"].unique()
        )
    }

    return summary


def get_sentiment_distribution(dataframe):
    """
    Generate sentiment category distribution.
    """
    sentiment_distribution = (
        dataframe["classification"]
        .value_counts()
        .reset_index()
    )

    sentiment_distribution.columns = [
        "Sentiment Category",
        "Count"
    ]

    return sentiment_distribution


def plot_sentiment_distribution(dataframe):
    """
    Plot sentiment category distribution.
    """
    sentiment_counts = dataframe["classification"].value_counts()

    plt.figure(figsize=(10, 5))
    sentiment_counts.plot(kind="bar")
    plt.title("Sentiment Category Distribution")
    plt.xlabel("Sentiment Category")
    plt.ylabel("Count")
    plt.tight_layout()
    plt.show()


def plot_sentiment_score_distribution(dataframe):
    """
    Plot sentiment score distribution.
    """
    plt.figure(figsize=(10, 5))
    plt.hist(dataframe["value"], bins=20)
    plt.title("Fear & Greed Score Distribution")
    plt.xlabel("Sentiment Score")
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

fear_greed_path = (
    "/content/drive/MyDrive/"
    "Primetrade_Assignment/fear_greed_index.csv"
)

fear_greed_df = load_fear_greed_dataset(
    fear_greed_path
)

fear_greed_df = preprocess_fear_greed_dataset(
    fear_greed_df
)

summary = get_fear_greed_summary(
    fear_greed_df
)

sentiment_distribution = (
    get_sentiment_distribution(
        fear_greed_df
    )
)

print("=" * 70)
print("FEAR & GREED DATASET SUMMARY")
print("=" * 70)

print("Rows:", summary["Rows"])
print("Columns:", summary["Columns"])

print("\nColumn Names:")
for column_name in summary["Column Names"]:
    print("-", column_name)

print("\nDate Range:")
print(summary["Date Range Start"].date(),
      "to",
      summary["Date Range End"].date())

print("\nSentiment Categories:")
for category in summary["Sentiment Categories"]:
    print("-", category)

print("\nData Types")
print(fear_greed_df.dtypes)

print("\nFirst 5 Rows")
display(fear_greed_df.head())

print("\nSentiment Distribution")
display(sentiment_distribution)

plot_sentiment_distribution(
    fear_greed_df
)

plot_sentiment_score_distribution(
    fear_greed_df
)
```

```python
# ============================================================
# DATASET 2: HISTORICAL TRADING DATASET UNDERSTANDING
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def load_historical_dataset(file_path):
    """
    Load historical trading dataset.
    """
    dataframe = pd.read_csv(file_path)

    return dataframe


def preprocess_historical_dataset(dataframe):
    """
    Convert timestamp column to datetime.
    """
    dataframe = dataframe.copy()

    dataframe["Timestamp IST"] = pd.to_datetime(
        dataframe["Timestamp IST"],
        dayfirst=True,
        errors="coerce"
    )

    return dataframe


def get_dataset_summary(dataframe):
    """
    Generate dataset summary statistics.
    """
    summary = {
        "Rows": dataframe.shape[0],
        "Columns": dataframe.shape[1],
        "Column Names": list(dataframe.columns),
        "Date Range Start": dataframe["Timestamp IST"].min(),
        "Date Range End": dataframe["Timestamp IST"].max(),
        "Unique Traders": dataframe["Account"].nunique(),
        "Unique Coins": dataframe["Coin"].nunique(),
        "Total Trades": len(dataframe)
    }

    return summary


def get_coin_distribution(dataframe, top_n=10):
    """
    Get top traded coins.
    """
    coin_distribution = (
        dataframe["Coin"]
        .value_counts()
        .head(top_n)
        .reset_index()
    )

    coin_distribution.columns = [
        "Coin",
        "Trade Count"
    ]

    return coin_distribution


def get_direction_distribution(dataframe):
    """
    Get Buy/Sell distribution.
    """
    direction_distribution = (
        dataframe["Direction"]
        .value_counts()
        .reset_index()
    )

    direction_distribution.columns = [
        "Direction",
        "Count"
    ]

    return direction_distribution


def get_pnl_summary(dataframe):
    """
    Get profit/loss statistics.
    """
    pnl_summary = dataframe["Closed PnL"].describe()

    return pnl_summary


def plot_top_traded_coins(dataframe, top_n=10):
    """
    Plot top traded coins.
    """
    top_coins = (
        dataframe["Coin"]
        .value_counts()
        .head(top_n)
    )

    plt.figure(figsize=(12, 6))
    top_coins.plot(kind="bar")

    plt.title(f"Top {top_n} Most Traded Coins")
    plt.xlabel("Coin")
    plt.ylabel("Number of Trades")

    plt.tight_layout()
    plt.show()


def plot_trade_direction_distribution(dataframe):
    """
    Plot Buy vs Sell distribution.
    """
    direction_counts = (
        dataframe["Direction"]
        .value_counts()
    )

    plt.figure(figsize=(7, 7))

    plt.pie(
        direction_counts,
        labels=direction_counts.index,
        autopct="%1.1f%%"
    )

    plt.title("Trade Direction Distribution")

    plt.show()


def plot_closed_pnl_distribution(dataframe):
    """
    Plot Closed PnL distribution.
    """
    plt.figure(figsize=(12, 6))

    plt.hist(
        dataframe["Closed PnL"],
        bins=50
    )

    plt.title("Closed PnL Distribution")
    plt.xlabel("Closed PnL")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

historical_data_path = (
    "/content/drive/MyDrive/"
    "Primetrade_Assignment/historical_data.csv"
)

historical_df = load_historical_dataset(
    historical_data_path
)

historical_df = preprocess_historical_dataset(
    historical_df
)

summary = get_dataset_summary(
    historical_df
)

coin_distribution = get_coin_distribution(
    historical_df
)

direction_distribution = (
    get_direction_distribution(
        historical_df
    )
)

pnl_summary = get_pnl_summary(
    historical_df
)

# ============================================================
# DISPLAY OUTPUTS
# ============================================================

print("=" * 70)
print("HISTORICAL TRADING DATASET SUMMARY")
print("=" * 70)

print(f"Rows: {summary['Rows']}")
print(f"Columns: {summary['Columns']}")

print("\nDate Range:")
print(summary["Date Range Start"])
print("to")
print(summary["Date Range End"])

print("\nUnique Traders:",
      summary["Unique Traders"])

print("Unique Coins:",
      summary["Unique Coins"])

print("Total Trades:",
      summary["Total Trades"])

print("\n" + "=" * 70)
print("COLUMN NAMES")
print("=" * 70)

for column_name in summary["Column Names"]:
    print("-", column_name)

print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)

print(historical_df.dtypes)

print("\n" + "=" * 70)
print("FIRST 5 ROWS")
print("=" * 70)

display(historical_df.head())

print("\n" + "=" * 70)
print("TOP TRADED COINS")
print("=" * 70)

display(coin_distribution)

print("\n" + "=" * 70)
print("BUY / SELL DISTRIBUTION")
print("=" * 70)

display(direction_distribution)

print("\n" + "=" * 70)
print("CLOSED PnL SUMMARY")
print("=" * 70)

print(pnl_summary)

# ============================================================
# PLOTS
# ============================================================

plot_top_traded_coins(
    historical_df
)

plot_trade_direction_distribution(
    historical_df
)

plot_closed_pnl_distribution(
    historical_df
)
```

IDENTIFY COMMON LINK BETWEEN DATASETS

```python
# ============================================================
# STEP 3: IDENTIFY COMMON LINK BETWEEN DATASETS
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def load_datasets(
    fear_greed_path,
    historical_data_path
):
    """
    Load both datasets.
    """

    fear_greed_df = pd.read_csv(
        fear_greed_path
    )

    historical_df = pd.read_csv(
        historical_data_path
    )

    return (
        fear_greed_df,
        historical_df
    )


def preprocess_datasets(
    fear_greed_df,
    historical_df
):
    """
    Create merge-ready date columns.
    """

    fear_greed_df = fear_greed_df.copy()
    historical_df = historical_df.copy()

    fear_greed_df["date"] = pd.to_datetime(
        fear_greed_df["date"]
    )

    historical_df["Timestamp IST"] = pd.to_datetime(
        historical_df["Timestamp IST"],
        dayfirst=True,
        errors="coerce"
    )

    fear_greed_df["Merge_Date"] = (
        fear_greed_df["date"].dt.date
    )

    historical_df["Merge_Date"] = (
        historical_df["Timestamp IST"].dt.date
    )

    return (
        fear_greed_df,
        historical_df
    )


def get_date_overlap_summary(
    fear_greed_df,
    historical_df
):
    """
    Check overlap between datasets.
    """

    sentiment_dates = set(
        fear_greed_df["Merge_Date"]
    )

    trading_dates = set(
        historical_df["Merge_Date"]
    )

    common_dates = (
        sentiment_dates &
        trading_dates
    )

    overlap_summary = {
        "Sentiment Dates":
            len(sentiment_dates),

        "Trading Dates":
            len(trading_dates),

        "Common Dates":
            len(common_dates),

        "Coverage Percentage":
            (
                len(common_dates)
                / len(trading_dates)
            ) * 100
    }

    return overlap_summary


def create_merged_preview(
    fear_greed_df,
    historical_df
):
    """
    Create sample merged dataset.
    """

    merged_df = historical_df.merge(
        fear_greed_df[
            [
                "Merge_Date",
                "value",
                "classification"
            ]
        ],
        on="Merge_Date",
        how="left"
    )

    return merged_df


def get_merge_statistics(
    merged_df
):
    """
    Calculate merge success.
    """

    matched_rows = (
        merged_df["classification"]
        .notna()
        .sum()
    )

    total_rows = len(
        merged_df
    )

    statistics = {
        "Total Trades":
            total_rows,

        "Matched Trades":
            matched_rows,

        "Unmatched Trades":
            total_rows - matched_rows,

        "Match Percentage":
            (
                matched_rows /
                total_rows
            ) * 100
    }

    return statistics


def plot_date_range_comparison(
    fear_greed_df,
    historical_df
):
    """
    Compare date ranges.
    """

    plt.figure(
        figsize=(12, 4)
    )

    plt.plot(
        [
            fear_greed_df["date"].min(),
            fear_greed_df["date"].max()
        ],
        [1, 1],
        linewidth=8,
        label="Fear & Greed"
    )

    plt.plot(
        [
            historical_df["Timestamp IST"].min(),
            historical_df["Timestamp IST"].max()
        ],
        [2, 2],
        linewidth=8,
        label="Trading Data"
    )

    plt.yticks(
        [1, 2],
        [
            "Fear & Greed",
            "Trading Data"
        ]
    )

    plt.title(
        "Date Range Comparison"
    )

    plt.legend()

    plt.tight_layout()
    plt.show()


def plot_sentiment_after_merge(
    merged_df
):
    """
    Sentiment distribution
    after merge.
    """

    sentiment_counts = (
        merged_df["classification"]
        .value_counts()
    )

    plt.figure(
        figsize=(10, 5)
    )

    sentiment_counts.plot(
        kind="bar"
    )

    plt.title(
        "Sentiment Distribution After Merge"
    )

    plt.xlabel(
        "Sentiment"
    )

    plt.ylabel(
        "Number of Trades"
    )

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

fear_greed_path = (
    "/content/drive/MyDrive/"
    "Primetrade_Assignment/"
    "fear_greed_index.csv"
)

historical_data_path = (
    "/content/drive/MyDrive/"
    "Primetrade_Assignment/"
    "historical_data.csv"
)

(
    fear_greed_df,
    historical_df
) = load_datasets(
    fear_greed_path,
    historical_data_path
)

(
    fear_greed_df,
    historical_df
) = preprocess_datasets(
    fear_greed_df,
    historical_df
)

overlap_summary = (
    get_date_overlap_summary(
        fear_greed_df,
        historical_df
    )
)

merged_df = (
    create_merged_preview(
        fear_greed_df,
        historical_df
    )
)

merge_statistics = (
    get_merge_statistics(
        merged_df
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("COMMON LINK IDENTIFICATION")
print("=" * 70)

print(
    "\nThe datasets will be merged using "
    "trade date and sentiment date "
    "to associate each trade with the "
    "prevailing market sentiment."
)

print("\nDate Overlap Summary")

for key, value in overlap_summary.items():
    print(
        f"{key}: {value}"
    )

print("\nMerge Statistics")

for key, value in merge_statistics.items():
    print(
        f"{key}: {value}"
    )

print("\nMerged Dataset Preview")

display(
    merged_df[
        [
            "Account",
            "Coin",
            "Closed PnL",
            "Merge_Date",
            "classification",
            "value"
        ]
    ].head()
)

# ============================================================
# PLOTS
# ============================================================

plot_date_range_comparison(
    fear_greed_df,
    historical_df
)

plot_sentiment_after_merge(
    merged_df
)
```

```python
# ============================================================
# STEP 4: IDENTIFY POSSIBLE FEATURES
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def get_profitability_summary(dataframe):
    """
    Generate profitability feature summary.
    """

    profitability_summary = (
        dataframe["Closed PnL"]
        .describe()
        .to_frame()
        .reset_index()
    )

    profitability_summary.columns = [
        "Statistic",
        "Closed PnL"
    ]

    return profitability_summary


def get_risk_summary(dataframe):
    """
    Generate risk feature summary.
    """

    risk_summary = (
        dataframe["Size USD"]
        .describe()
        .to_frame()
        .reset_index()
    )

    risk_summary.columns = [
        "Statistic",
        "Size USD"
    ]

    return risk_summary


def get_activity_summary(dataframe):
    """
    Generate trading activity summary.
    """

    activity_summary = {
        "Total Trades":
            len(dataframe),

        "Unique Traders":
            dataframe["Account"].nunique(),

        "Unique Coins":
            dataframe["Coin"].nunique()
    }

    return activity_summary


def get_direction_summary(dataframe):
    """
    Generate trade direction summary.
    """

    direction_summary = (
        dataframe["Direction"]
        .value_counts()
        .reset_index()
    )

    direction_summary.columns = [
        "Direction",
        "Count"
    ]

    return direction_summary


def plot_closed_pnl_distribution(dataframe):
    """
    Plot profitability distribution.
    """

    plt.figure(figsize=(12, 6))

    plt.hist(
        dataframe["Closed PnL"],
        bins=100
    )

    plt.title(
        "Profitability Feature: Closed PnL Distribution"
    )

    plt.xlabel("Closed PnL")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


def plot_size_usd_distribution(dataframe):
    """
    Plot position size distribution.
    """

    plt.figure(figsize=(12, 6))

    plt.hist(
        dataframe["Size USD"],
        bins=100
    )

    plt.title(
        "Risk Feature: Size USD Distribution"
    )

    plt.xlabel("Position Size (USD)")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


def plot_direction_distribution(dataframe):
    """
    Plot trade direction distribution.
    """

    direction_counts = (
        dataframe["Direction"]
        .value_counts()
        .head(10)
    )

    plt.figure(figsize=(12, 6))

    direction_counts.plot(
        kind="bar"
    )

    plt.title(
        "Direction Feature Distribution"
    )

    plt.xlabel("Direction")
    plt.ylabel("Count")

    plt.tight_layout()
    plt.show()


def plot_daily_trade_activity(dataframe):
    """
    Plot number of trades per day.
    """

    daily_trades = (
        dataframe
        .groupby("Merge_Date")
        .size()
    )

    plt.figure(figsize=(14, 6))

    plt.plot(
        daily_trades.index,
        daily_trades.values
    )

    plt.title(
        "Activity Feature: Daily Number of Trades"
    )

    plt.xlabel("Date")
    plt.ylabel("Number of Trades")

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

# merged_df is already available from Step 3

profitability_summary = (
    get_profitability_summary(
        merged_df
    )
)

risk_summary = (
    get_risk_summary(
        merged_df
    )
)

activity_summary = (
    get_activity_summary(
        merged_df
    )
)

direction_summary = (
    get_direction_summary(
        merged_df
    )
)

# ============================================================
# DISPLAY OUTPUTS
# ============================================================

print("=" * 70)
print("PROFITABILITY FEATURE")
print("=" * 70)

print("\nFeature Used:")
print("- Closed PnL")

print("\nBusiness Questions:")
print("- Do traders earn more during Greed?")
print("- Do traders lose more during Fear?")

display(profitability_summary)

print("\n")

print("=" * 70)
print("RISK FEATURE")
print("=" * 70)

print("\nFeature Used:")
print("- Size USD")

print("\nBusiness Questions:")
print("- Are traders taking more risk during Greed?")
print("- Do they reduce risk during Fear?")

display(risk_summary)

print("\n")

print("=" * 70)
print("ACTIVITY FEATURE")
print("=" * 70)

print("\nFeature Used:")
print("- Number of Trades")

print("\nBusiness Questions:")
print("- Is trading activity higher during Greed?")

for key, value in activity_summary.items():
    print(f"{key}: {value}")

print("\n")

print("=" * 70)
print("DIRECTION FEATURE")
print("=" * 70)

print("\nFeature Used:")
print("- Direction")

print("\nBusiness Questions:")
print("- Do traders go long more often during Greed?")

display(direction_summary)

# ============================================================
# PLOTS
# ============================================================

plot_closed_pnl_distribution(
    merged_df
)

plot_size_usd_distribution(
    merged_df
)

plot_direction_distribution(
    merged_df
)

plot_daily_trade_activity(
    merged_df
)
```

```python
# ============================================================
# STEP 5: HYPOTHESIS VALIDATION
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def test_h1_risk_behavior(dataframe):
    """
    H1:
    Traders take larger position sizes during Greed.
    """

    risk_analysis = (
        dataframe
        .groupby("classification")["Size USD"]
        .agg(
            [
                "count",
                "mean",
                "median",
                "max"
            ]
        )
        .reset_index()
    )

    return risk_analysis


def test_h2_trading_activity(dataframe):
    """
    H2:
    Trading activity increases during Greed.
    """

    activity_analysis = (
        dataframe
        .groupby("classification")
        .size()
        .reset_index(name="Trade Count")
        .sort_values(
            by="Trade Count",
            ascending=False
        )
    )

    return activity_analysis


def test_h3_profitability(dataframe):
    """
    H3:
    Profitability differs across sentiment categories.
    """

    profitability_analysis = (
        dataframe
        .groupby("classification")["Closed PnL"]
        .agg(
            [
                "count",
                "mean",
                "median",
                "sum",
                "std"
            ]
        )
        .reset_index()
    )

    return profitability_analysis


def test_h4_volatility(dataframe):
    """
    H4:
    Extreme Fear and Extreme Greed
    create higher volatility.
    """

    volatility_analysis = (
        dataframe
        .groupby("classification")["Closed PnL"]
        .std()
        .reset_index()
    )

    volatility_analysis.columns = [
        "Sentiment",
        "PnL Volatility"
    ]

    volatility_analysis = (
        volatility_analysis
        .sort_values(
            by="PnL Volatility",
            ascending=False
        )
    )

    return volatility_analysis


def plot_h1_risk_behavior(risk_analysis):
    """
    Average Size USD by sentiment.
    """

    plt.figure(figsize=(10, 5))

    plt.bar(
        risk_analysis["classification"],
        risk_analysis["mean"]
    )

    plt.title(
        "H1: Average Position Size by Sentiment"
    )

    plt.xlabel(
        "Sentiment"
    )

    plt.ylabel(
        "Average Size USD"
    )

    plt.xticks(rotation=20)

    plt.tight_layout()
    plt.show()


def plot_h2_trading_activity(activity_analysis):
    """
    Trade count by sentiment.
    """

    plt.figure(figsize=(10, 5))

    plt.bar(
        activity_analysis["classification"],
        activity_analysis["Trade Count"]
    )

    plt.title(
        "H2: Trading Activity by Sentiment"
    )

    plt.xlabel(
        "Sentiment"
    )

    plt.ylabel(
        "Number of Trades"
    )

    plt.xticks(rotation=20)

    plt.tight_layout()
    plt.show()


def plot_h3_profitability(profitability_analysis):
    """
    Mean PnL by sentiment.
    """

    plt.figure(figsize=(10, 5))

    plt.bar(
        profitability_analysis["classification"],
        profitability_analysis["mean"]
    )

    plt.title(
        "H3: Average Closed PnL by Sentiment"
    )

    plt.xlabel(
        "Sentiment"
    )

    plt.ylabel(
        "Average Closed PnL"
    )

    plt.xticks(rotation=20)

    plt.tight_layout()
    plt.show()


def plot_h4_volatility(volatility_analysis):
    """
    PnL volatility by sentiment.
    """

    plt.figure(figsize=(10, 5))

    plt.bar(
        volatility_analysis["Sentiment"],
        volatility_analysis["PnL Volatility"]
    )

    plt.title(
        "H4: Trading Volatility by Sentiment"
    )

    plt.xlabel(
        "Sentiment"
    )

    plt.ylabel(
        "PnL Standard Deviation"
    )

    plt.xticks(rotation=20)

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

risk_analysis = test_h1_risk_behavior(
    merged_df
)

activity_analysis = test_h2_trading_activity(
    merged_df
)

profitability_analysis = test_h3_profitability(
    merged_df
)

volatility_analysis = test_h4_volatility(
    merged_df
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("H1: POSITION SIZE VS SENTIMENT")
print("=" * 70)

display(risk_analysis)

print("\n")

print("=" * 70)
print("H2: TRADING ACTIVITY VS SENTIMENT")
print("=" * 70)

display(activity_analysis)

print("\n")

print("=" * 70)
print("H3: PROFITABILITY VS SENTIMENT")
print("=" * 70)

display(profitability_analysis)

print("\n")

print("=" * 70)
print("H4: VOLATILITY VS SENTIMENT")
print("=" * 70)

display(volatility_analysis)

# ============================================================
# PLOTS
# ============================================================

plot_h1_risk_behavior(
    risk_analysis
)

plot_h2_trading_activity(
    activity_analysis
)

plot_h3_profitability(
    profitability_analysis
)

plot_h4_volatility(
    volatility_analysis
)
```

# EXPLORATORY DATA ANALYSIS (EDA)

Load Final Dataset

```python
# ============================================================
# PHASE 3 - STEP 1: LOAD FINAL DATASET
# Objective:
# Load the final EDA dataset and inspect its structure.
# Outputs:
# - Number of rows
# - Number of columns
# - Column names
# - Data types
# - First 5 rows
# ============================================================

import pandas as pd


def load_dataset(file_path):
    """
    Load the final EDA dataset.
    """
    dataframe = pd.read_csv(file_path)
    return dataframe


def get_dataset_overview(dataframe):
    """
    Generate basic dataset information.
    """
    dataset_info = {
        "Rows": dataframe.shape[0],
        "Columns": dataframe.shape[1],
        "Column Names": list(dataframe.columns),
        "Data Types": dataframe.dtypes
    }
    return dataset_info


def get_first_records(dataframe, number_of_rows=5):
    """
    Return the first few records of the dataset.
    """
    return dataframe.head(number_of_rows)


# ============================================================
# MAIN PROGRAM
# ============================================================

final_dataset_path = (
    "/content/drive/MyDrive/Primetrade_Assignment/"
    "final_eda_dataset.csv"
)

# Load dataset
final_eda_df = load_dataset(final_dataset_path)

# Get overview information
dataset_overview = get_dataset_overview(final_eda_df)

# Get first 5 rows
first_five_rows = get_first_records(final_eda_df)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 3 - STEP 1: LOAD FINAL DATASET")
print("=" * 70)

print(f"Rows: {dataset_overview['Rows']}")
print(f"Columns: {dataset_overview['Columns']}")

print("\n" + "=" * 70)
print("COLUMN NAMES")
print("=" * 70)

for column_name in dataset_overview["Column Names"]:
    print("-", column_name)

print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)

print(dataset_overview["Data Types"])

print("\n" + "=" * 70)
print("FIRST 5 ROWS")
print("=" * 70)

display(first_five_rows)
```

DATASET OVERVIEW.

```python
# ============================================================
# PHASE 3 - STEP 2: DATASET OVERVIEW
# Objective:
# Understand overall dataset characteristics.
#
# Outputs:
# - Total Trades
# - Unique Traders
# - Unique Coins
# - Date Range
# - Sentiment Categories
# - Dataset Overview Visualizations
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def preprocess_dataset(dataframe):
    """
    Convert Merge_Date to datetime format.
    """
    dataframe = dataframe.copy()

    dataframe["Merge_Date"] = pd.to_datetime(
        dataframe["Merge_Date"],
        errors="coerce"
    )

    return dataframe


def get_dataset_overview(dataframe):
    """
    Generate overall dataset statistics.
    """
    overview = {
        "Total Trades": len(dataframe),
        "Unique Traders": dataframe["Account"].nunique(),
        "Unique Coins": dataframe["Coin"].nunique(),
        "Date Range Start": dataframe["Merge_Date"].min(),
        "Date Range End": dataframe["Merge_Date"].max(),
        "Sentiment Categories": sorted(
            dataframe["classification"].dropna().unique()
        )
    }

    return overview


def get_sentiment_counts(dataframe):
    """
    Generate sentiment category counts.
    """
    sentiment_counts = (
        dataframe["classification"]
        .value_counts()
        .reset_index()
    )

    sentiment_counts.columns = [
        "Sentiment Category",
        "Trade Count"
    ]

    return sentiment_counts


def plot_dataset_overview(overview):
    """
    Plot high-level dataset statistics.
    """
    metric_names = [
        "Total Trades",
        "Unique Traders",
        "Unique Coins"
    ]

    metric_values = [
        overview["Total Trades"],
        overview["Unique Traders"],
        overview["Unique Coins"]
    ]

    plt.figure(figsize=(8, 5))
    plt.bar(metric_names, metric_values)

    plt.title("Dataset Overview Metrics")
    plt.ylabel("Count")
    plt.xlabel("Metric")

    plt.tight_layout()
    plt.show()


def plot_sentiment_categories(dataframe):
    """
    Plot sentiment category distribution.
    """
    sentiment_counts = (
        dataframe["classification"]
        .value_counts()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(8, 5))
    sentiment_counts.plot(kind="bar")

    plt.title("Available Sentiment Categories")
    plt.xlabel("Sentiment")
    plt.ylabel("Number of Trades")

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

# final_eda_df is already available from STEP 1

final_eda_df = preprocess_dataset(final_eda_df)

dataset_overview = get_dataset_overview(
    final_eda_df
)

sentiment_counts = get_sentiment_counts(
    final_eda_df
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 3 - STEP 2: DATASET OVERVIEW")
print("=" * 70)

print(f"Total Trades   : {dataset_overview['Total Trades']:,}")
print(f"Unique Traders : {dataset_overview['Unique Traders']:,}")
print(f"Unique Coins   : {dataset_overview['Unique Coins']:,}")

print("\n" + "=" * 70)
print("DATE RANGE")
print("=" * 70)

print(
    dataset_overview["Date Range Start"].date(),
    "to",
    dataset_overview["Date Range End"].date()
)

print("\n" + "=" * 70)
print("SENTIMENT CATEGORIES")
print("=" * 70)

for sentiment in dataset_overview["Sentiment Categories"]:
    print("-", sentiment)

print("\n" + "=" * 70)
print("SENTIMENT CATEGORY COUNTS")
print("=" * 70)

display(sentiment_counts)

# ============================================================
# BUSINESS UNDERSTANDING
# ============================================================

print("\n" + "=" * 70)
print("BUSINESS UNDERSTANDING")
print("=" * 70)

print(
    f"1. The dataset contains "
    f"{dataset_overview['Total Trades']:,} trading records."
)

print(
    f"2. Trading activity comes from "
    f"{dataset_overview['Unique Traders']} unique traders."
)

print(
    f"3. Traders interacted with "
    f"{dataset_overview['Unique Coins']} different coins."
)

print(
    f"4. The dataset spans from "
    f"{dataset_overview['Date Range Start'].date()} "
    f"to "
    f"{dataset_overview['Date Range End'].date()}."
)

print(
    f"5. Market sentiment is categorized into "
    f"{len(dataset_overview['Sentiment Categories'])} sentiment classes."
)

# ============================================================
# PLOTS
# ============================================================

plot_dataset_overview(
    dataset_overview
)

plot_sentiment_categories(
    final_eda_df
)
```

Sentiment Distribution Analysis

```python
# ============================================================
# PHASE 3 - STEP 3: SENTIMENT DISTRIBUTION ANALYSIS
# Objective:
# Analyze the distribution of market sentiment in the dataset.
#
# Features:
# - classification
# - value
#
# Outputs:
# - Sentiment counts
# - Sentiment percentages
# - Count Plot
# - Pie Chart
# - Fear & Greed Score Histogram
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def get_sentiment_counts(dataframe):
    """
    Generate sentiment frequency counts.
    """
    sentiment_counts = (
        dataframe["classification"]
        .value_counts()
        .reset_index()
    )

    sentiment_counts.columns = [
        "Sentiment",
        "Count"
    ]

    return sentiment_counts


def get_sentiment_percentages(dataframe):
    """
    Generate sentiment percentage distribution.
    """
    sentiment_percentages = (
        dataframe["classification"]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
        .reset_index()
    )

    sentiment_percentages.columns = [
        "Sentiment",
        "Percentage"
    ]

    return sentiment_percentages


def get_sentiment_summary(dataframe):
    """
    Generate combined sentiment summary.
    """
    sentiment_summary = (
        dataframe["classification"]
        .value_counts()
        .reset_index()
    )

    sentiment_summary.columns = [
        "Sentiment",
        "Count"
    ]

    sentiment_summary["Percentage"] = (
        sentiment_summary["Count"]
        / sentiment_summary["Count"].sum()
        * 100
    ).round(2)

    return sentiment_summary


def plot_sentiment_count(dataframe):
    """
    Plot sentiment count distribution.
    """
    sentiment_counts = (
        dataframe["classification"]
        .value_counts()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(8, 5))

    sentiment_counts.plot(
        kind="bar"
    )

    plt.title("Sentiment Distribution by Trade Count")
    plt.xlabel("Sentiment Category")
    plt.ylabel("Number of Trades")

    plt.tight_layout()
    plt.show()


def plot_sentiment_pie_chart(dataframe):
    """
    Plot sentiment percentage distribution.
    """
    sentiment_counts = (
        dataframe["classification"]
        .value_counts()
    )

    plt.figure(figsize=(8, 8))

    plt.pie(
        sentiment_counts,
        labels=sentiment_counts.index,
        autopct="%1.1f%%"
    )

    plt.title("Sentiment Percentage Distribution")

    plt.tight_layout()
    plt.show()


def plot_fear_greed_histogram(dataframe):
    """
    Plot Fear & Greed score distribution.
    """
    plt.figure(figsize=(10, 5))

    plt.hist(
        dataframe["value"],
        bins=30
    )

    plt.title("Fear & Greed Score Distribution")
    plt.xlabel("Fear & Greed Score")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


def identify_dominant_sentiment(sentiment_summary):
    """
    Identify the most frequent sentiment category.
    """
    dominant_sentiment = sentiment_summary.iloc[0]["Sentiment"]
    dominant_percentage = sentiment_summary.iloc[0]["Percentage"]

    return dominant_sentiment, dominant_percentage


# ============================================================
# MAIN PROGRAM
# ============================================================

# final_eda_df is already available from previous steps

sentiment_counts = get_sentiment_counts(
    final_eda_df
)

sentiment_percentages = get_sentiment_percentages(
    final_eda_df
)

sentiment_summary = get_sentiment_summary(
    final_eda_df
)

dominant_sentiment, dominant_percentage = (
    identify_dominant_sentiment(
        sentiment_summary
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 3 - STEP 3: SENTIMENT DISTRIBUTION ANALYSIS")
print("=" * 70)

print("\nSentiment Counts")
print("-" * 70)

display(sentiment_counts)

print("\nSentiment Percentages")
print("-" * 70)

display(sentiment_percentages)

print("\nCombined Sentiment Summary")
print("-" * 70)

display(sentiment_summary)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("INSIGHTS")
print("=" * 70)

print(
    f"Dominant Sentiment: {dominant_sentiment}"
)

print(
    f"Percentage of Trades: {dominant_percentage:.2f}%"
)

largest_percentage = sentiment_summary["Percentage"].max()
smallest_percentage = sentiment_summary["Percentage"].min()

if (largest_percentage - smallest_percentage) <= 10:
    balance_comment = "The sentiment distribution appears relatively balanced."
else:
    balance_comment = (
        "The sentiment distribution is not balanced and some "
        "sentiment categories occur much more frequently."
    )

print(
    f"Dataset Balance Assessment: {balance_comment}"
)

# ============================================================
# PLOTS
# ============================================================

plot_sentiment_count(
    final_eda_df
)

plot_sentiment_pie_chart(
    final_eda_df
)

plot_fear_greed_histogram(
    final_eda_df
)
```

Profitability Analysis (Closed PnL)

```python
# ============================================================
# PHASE 3 - STEP 4: PROFITABILITY ANALYSIS
# Objective:
# Analyze the distribution of Closed PnL.
#
# Feature:
# - Closed PnL
#
# Outputs:
# - Summary Statistics
# - Histogram
# - KDE Plot
# - Boxplot
#
# Insights:
# - Are profits concentrated?
# - Are losses extreme?
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def get_pnl_statistics(dataframe):
    """
    Generate summary statistics for Closed PnL.
    """
    pnl_statistics = dataframe["Closed PnL"].describe()

    return pnl_statistics


def get_profit_loss_summary(dataframe):
    """
    Calculate profit and loss trade counts.
    """
    profit_trades = (dataframe["Closed PnL"] > 0).sum()
    loss_trades = (dataframe["Closed PnL"] < 0).sum()
    breakeven_trades = (dataframe["Closed PnL"] == 0).sum()

    summary = pd.DataFrame({
        "Category": ["Profit Trades", "Loss Trades", "Breakeven Trades"],
        "Count": [profit_trades, loss_trades, breakeven_trades]
    })

    return summary


def calculate_pnl_percentiles(dataframe):
    """
    Calculate important PnL percentiles.
    """
    percentile_values = dataframe["Closed PnL"].quantile(
        [0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
    )

    percentile_dataframe = percentile_values.reset_index()
    percentile_dataframe.columns = [
        "Percentile",
        "Closed PnL"
    ]

    return percentile_dataframe


def plot_pnl_histogram(dataframe):
    """
    Plot Closed PnL histogram.
    """
    plt.figure(figsize=(10, 5))

    plt.hist(
        dataframe["Closed PnL"],
        bins=50
    )

    plt.title("Distribution of Closed PnL")
    plt.xlabel("Closed PnL")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


def plot_pnl_kde(dataframe):
    """
    Plot KDE distribution of Closed PnL.
    """
    plt.figure(figsize=(10, 5))

    dataframe["Closed PnL"].plot(
        kind="kde"
    )

    plt.title("KDE Plot of Closed PnL")
    plt.xlabel("Closed PnL")

    plt.tight_layout()
    plt.show()


def plot_pnl_boxplot(dataframe):
    """
    Plot Closed PnL boxplot.
    """
    plt.figure(figsize=(10, 2))

    plt.boxplot(
        dataframe["Closed PnL"],
        vert=False
    )

    plt.title("Closed PnL Boxplot")

    plt.tight_layout()
    plt.show()


def identify_extreme_trades(dataframe):
    """
    Identify largest profit and largest loss.
    """
    largest_profit = dataframe["Closed PnL"].max()
    largest_loss = dataframe["Closed PnL"].min()

    return largest_profit, largest_loss


# ============================================================
# MAIN PROGRAM
# ============================================================

pnl_statistics = get_pnl_statistics(
    final_eda_df
)

profit_loss_summary = get_profit_loss_summary(
    final_eda_df
)

pnl_percentiles = calculate_pnl_percentiles(
    final_eda_df
)

largest_profit, largest_loss = (
    identify_extreme_trades(
        final_eda_df
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 3 - STEP 4: PROFITABILITY ANALYSIS")
print("=" * 70)

print("\nSUMMARY STATISTICS")
print("-" * 70)

print(pnl_statistics)

print("\n")
print("=" * 70)
print("PROFIT / LOSS TRADE COUNTS")
print("=" * 70)

display(profit_loss_summary)

print("\n")
print("=" * 70)
print("PNL PERCENTILES")
print("=" * 70)

display(pnl_percentiles)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("PROFITABILITY INSIGHTS")
print("=" * 70)

print(
    f"Largest Profit Observed : {largest_profit:,.2f}"
)

print(
    f"Largest Loss Observed   : {largest_loss:,.2f}"
)

if abs(largest_loss) > largest_profit:
    print(
        "Observation: Extreme losses are larger than extreme profits."
    )
else:
    print(
        "Observation: Extreme profits are larger than extreme losses."
    )

mean_pnl = final_eda_df["Closed PnL"].mean()
median_pnl = final_eda_df["Closed PnL"].median()

print(
    f"Mean Closed PnL   : {mean_pnl:,.2f}"
)

print(
    f"Median Closed PnL : {median_pnl:,.2f}"
)

if abs(mean_pnl - median_pnl) > (0.25 * max(abs(mean_pnl), abs(median_pnl), 1)):
    print(
        "Observation: Profits appear concentrated in a relatively small number of trades."
    )
else:
    print(
        "Observation: Profitability appears more evenly distributed."
    )

# ============================================================
# VISUALIZATIONS
# ============================================================

plot_pnl_histogram(
    final_eda_df
)

plot_pnl_kde(
    final_eda_df
)

plot_pnl_boxplot(
    final_eda_df
)
```

POSITION SIZE ANALYSIS

```python
# ============================================================
# PHASE 3 - STEP 5: POSITION SIZE ANALYSIS
# Objective:
# Analyze the distribution of trade position sizes.
#
# Feature:
# - Size USD
#
# Outputs:
# - Summary Statistics
# - Histogram
# - Boxplot
# - Log-Scale Histogram
#
# Insights:
# - Typical trade size
# - Presence of whale trades
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np


def get_position_size_statistics(dataframe):
    """
    Generate summary statistics for Size USD.
    """
    size_statistics = dataframe["Size USD"].describe()

    return size_statistics


def calculate_position_size_percentiles(dataframe):
    """
    Calculate important position size percentiles.
    """
    percentile_values = dataframe["Size USD"].quantile(
        [0.01, 0.05, 0.25, 0.50, 0.75, 0.90, 0.95, 0.99]
    )

    percentile_dataframe = percentile_values.reset_index()

    percentile_dataframe.columns = [
        "Percentile",
        "Size USD"
    ]

    return percentile_dataframe


def get_largest_positions(
    dataframe,
    top_n=20
):
    """
    Return largest position sizes.
    """
    largest_positions = (
        dataframe
        .sort_values(
            by="Size USD",
            ascending=False
        )
        .head(top_n)
    )

    return largest_positions


def calculate_iqr_outliers(dataframe):
    """
    Detect position size outliers using IQR.
    """
    q1 = dataframe["Size USD"].quantile(0.25)
    q3 = dataframe["Size USD"].quantile(0.75)

    iqr = q3 - q1

    lower_bound = q1 - (1.5 * iqr)
    upper_bound = q3 + (1.5 * iqr)

    outlier_count = (
        (
            dataframe["Size USD"] < lower_bound
        )
        |
        (
            dataframe["Size USD"] > upper_bound
        )
    ).sum()

    outlier_percentage = (
        outlier_count / len(dataframe)
    ) * 100

    summary = {
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier Percentage": outlier_percentage
    }

    return summary


def plot_size_histogram(dataframe):
    """
    Plot Size USD histogram.
    """
    plt.figure(figsize=(10, 5))

    plt.hist(
        dataframe["Size USD"],
        bins=50
    )

    plt.title("Position Size Distribution")
    plt.xlabel("Size USD")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


def plot_size_boxplot(dataframe):
    """
    Plot Size USD boxplot.
    """
    plt.figure(figsize=(10, 2))

    plt.boxplot(
        dataframe["Size USD"],
        vert=False
    )

    plt.title("Position Size Boxplot")

    plt.tight_layout()
    plt.show()


def plot_log_scale_histogram(dataframe):
    """
    Plot log-scale histogram for Size USD.
    """
    positive_sizes = dataframe[
        dataframe["Size USD"] > 0
    ]["Size USD"]

    plt.figure(figsize=(10, 5))

    plt.hist(
        np.log10(positive_sizes),
        bins=50
    )

    plt.title(
        "Log-Scale Position Size Distribution"
    )

    plt.xlabel("Log10(Size USD)")
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

size_statistics = get_position_size_statistics(
    final_eda_df
)

position_percentiles = (
    calculate_position_size_percentiles(
        final_eda_df
    )
)

largest_positions = get_largest_positions(
    final_eda_df
)

outlier_summary = calculate_iqr_outliers(
    final_eda_df
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 3 - STEP 5: POSITION SIZE ANALYSIS")
print("=" * 70)

print("\nSUMMARY STATISTICS")
print("-" * 70)

print(size_statistics)

print("\n")
print("=" * 70)
print("POSITION SIZE PERCENTILES")
print("=" * 70)

display(position_percentiles)

print("\n")
print("=" * 70)
print("TOP 20 LARGEST POSITIONS")
print("=" * 70)

display(
    largest_positions[
        [
            "Account",
            "Coin",
            "Size USD",
            "Closed PnL"
        ]
    ]
)

print("\n")
print("=" * 70)
print("OUTLIER SUMMARY")
print("=" * 70)

for key, value in outlier_summary.items():
    print(f"{key}: {value}")

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("POSITION SIZE INSIGHTS")
print("=" * 70)

median_position_size = (
    final_eda_df["Size USD"]
    .median()
)

mean_position_size = (
    final_eda_df["Size USD"]
    .mean()
)

maximum_position_size = (
    final_eda_df["Size USD"]
    .max()
)

print(
    f"Typical Trade Size (Median): "
    f"${median_position_size:,.2f}"
)

print(
    f"Average Trade Size: "
    f"${mean_position_size:,.2f}"
)

print(
    f"Largest Position Observed: "
    f"${maximum_position_size:,.2f}"
)

if maximum_position_size > (
    median_position_size * 100
):
    print(
        "Observation: Whale trades are present in the dataset."
    )
else:
    print(
        "Observation: Position sizes are relatively uniform."
    )

if mean_position_size > (
    median_position_size * 2
):
    print(
        "Observation: Position size distribution is heavily right-skewed."
    )
else:
    print(
        "Observation: Position size distribution is relatively balanced."
    )

# ============================================================
# VISUALIZATIONS
# ============================================================

plot_size_histogram(
    final_eda_df
)

plot_size_boxplot(
    final_eda_df
)

plot_log_scale_histogram(
    final_eda_df
)
```

TRADE DIRECTION ANALYSIS


```python
# ============================================================
# PHASE 3 - STEP 6: TRADE DIRECTION ANALYSIS
# Objective:
# Analyze trader direction behavior.
#
# Feature:
# - Direction
#
# Outputs:
# - Direction Counts
# - Direction Percentages
# - Bar Chart
# - Pie Chart
#
# Insights:
# - Long Bias?
# - Short Bias?
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def get_direction_counts(dataframe):
    """
    Calculate direction counts.
    """
    direction_counts = (
        dataframe["Direction"]
        .value_counts()
        .reset_index()
    )

    direction_counts.columns = [
        "Direction",
        "Count"
    ]

    return direction_counts


def get_direction_percentages(dataframe):
    """
    Calculate direction percentages.
    """
    direction_percentages = (
        dataframe["Direction"]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
        .reset_index()
    )

    direction_percentages.columns = [
        "Direction",
        "Percentage"
    ]

    return direction_percentages


def create_direction_summary(dataframe):
    """
    Create combined direction summary.
    """
    direction_summary = (
        dataframe["Direction"]
        .value_counts()
        .reset_index()
    )

    direction_summary.columns = [
        "Direction",
        "Count"
    ]

    direction_summary["Percentage"] = (
        direction_summary["Count"]
        / direction_summary["Count"].sum()
        * 100
    ).round(2)

    return direction_summary


def get_long_short_statistics(dataframe):
    """
    Calculate Long and Short related statistics.
    """
    long_trades = dataframe[
        dataframe["Direction"]
        .str.contains(
            "Long",
            case=False,
            na=False
        )
    ]

    short_trades = dataframe[
        dataframe["Direction"]
        .str.contains(
            "Short",
            case=False,
            na=False
        )
    ]

    statistics = {
        "Long Trades": len(long_trades),
        "Short Trades": len(short_trades)
    }

    return statistics


def plot_direction_bar_chart(dataframe):
    """
    Plot direction distribution.
    """
    direction_counts = (
        dataframe["Direction"]
        .value_counts()
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(12, 6))

    direction_counts.plot(
        kind="bar"
    )

    plt.title("Trade Direction Distribution")
    plt.xlabel("Direction")
    plt.ylabel("Number of Trades")

    plt.tight_layout()
    plt.show()


def plot_direction_pie_chart(dataframe):
    """
    Plot direction percentage distribution.
    """
    direction_counts = (
        dataframe["Direction"]
        .value_counts()
    )

    plt.figure(figsize=(8, 8))

    plt.pie(
        direction_counts,
        labels=direction_counts.index,
        autopct="%1.1f%%"
    )

    plt.title("Trade Direction Percentage Distribution")

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

direction_counts = get_direction_counts(
    final_eda_df
)

direction_percentages = (
    get_direction_percentages(
        final_eda_df
    )
)

direction_summary = (
    create_direction_summary(
        final_eda_df
    )
)

long_short_statistics = (
    get_long_short_statistics(
        final_eda_df
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 3 - STEP 6: TRADE DIRECTION ANALYSIS")
print("=" * 70)

print("\nDIRECTION COUNTS")
print("-" * 70)

display(direction_counts)

print("\nDIRECTION PERCENTAGES")
print("-" * 70)

display(direction_percentages)

print("\nCOMBINED DIRECTION SUMMARY")
print("-" * 70)

display(direction_summary)

print("\n")
print("=" * 70)
print("LONG VS SHORT STATISTICS")
print("=" * 70)

for key, value in long_short_statistics.items():
    print(f"{key}: {value:,}")

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("TRADE DIRECTION INSIGHTS")
print("=" * 70)

long_trade_count = (
    long_short_statistics["Long Trades"]
)

short_trade_count = (
    long_short_statistics["Short Trades"]
)

total_long_short = (
    long_trade_count +
    short_trade_count
)

if total_long_short > 0:

    long_percentage = (
        long_trade_count
        / total_long_short
    ) * 100

    short_percentage = (
        short_trade_count
        / total_long_short
    ) * 100

    print(
        f"Long Trade Percentage : "
        f"{long_percentage:.2f}%"
    )

    print(
        f"Short Trade Percentage: "
        f"{short_percentage:.2f}%"
    )

    if long_trade_count > short_trade_count:
        print(
            "Observation: Traders exhibit a LONG bias."
        )

    elif short_trade_count > long_trade_count:
        print(
            "Observation: Traders exhibit a SHORT bias."
        )

    else:
        print(
            "Observation: Long and Short activity is balanced."
        )

else:
    print(
        "No Long/Short specific directions found."
    )

most_common_direction = (
    direction_summary.iloc[0]["Direction"]
)

most_common_percentage = (
    direction_summary.iloc[0]["Percentage"]
)

print(
    f"\nMost Common Direction: "
    f"{most_common_direction}"
)

print(
    f"Share of Total Trades: "
    f"{most_common_percentage:.2f}%"
)

# ============================================================
# VISUALIZATIONS
# ============================================================

plot_direction_bar_chart(
    final_eda_df
)

plot_direction_pie_chart(
    final_eda_df
)
```

COIN ANALYSIS

```python
# ============================================================
# PHASE 3 - STEP 7: COIN ANALYSIS
# Objective:
# Analyze trading activity across different coins.
#
# Feature:
# - Coin
#
# Outputs:
# - Top 20 Coins
# - Top 20 Coins Bar Chart
#
# Insights:
# - Which assets dominate trading activity?
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def get_coin_counts(dataframe):
    """
    Calculate trade counts for all coins.
    """
    coin_counts = (
        dataframe["Coin"]
        .value_counts()
        .reset_index()
    )

    coin_counts.columns = [
        "Coin",
        "Trade Count"
    ]

    return coin_counts


def get_top_coins(
    dataframe,
    top_n=20
):
    """
    Get top traded coins.
    """
    top_coins = (
        dataframe["Coin"]
        .value_counts()
        .head(top_n)
        .reset_index()
    )

    top_coins.columns = [
        "Coin",
        "Trade Count"
    ]

    return top_coins


def calculate_coin_percentages(
    dataframe,
    top_n=20
):
    """
    Calculate percentage contribution of top coins.
    """
    total_trades = len(dataframe)

    coin_summary = (
        dataframe["Coin"]
        .value_counts()
        .head(top_n)
        .reset_index()
    )

    coin_summary.columns = [
        "Coin",
        "Trade Count"
    ]

    coin_summary["Percentage"] = (
        coin_summary["Trade Count"]
        / total_trades
        * 100
    ).round(2)

    return coin_summary


def calculate_top_coin_concentration(
    dataframe,
    top_n=3
):
    """
    Calculate concentration of top coins.
    """
    top_trade_count = (
        dataframe["Coin"]
        .value_counts()
        .head(top_n)
        .sum()
    )

    concentration_percentage = (
        top_trade_count
        / len(dataframe)
        * 100
    )

    return concentration_percentage


def plot_top_coins(
    dataframe,
    top_n=20
):
    """
    Plot top traded coins.
    """
    top_coins = (
        dataframe["Coin"]
        .value_counts()
        .head(top_n)
    )

    plt.figure(figsize=(14, 6))

    top_coins.plot(
        kind="bar"
    )

    plt.title(
        f"Top {top_n} Most Traded Coins"
    )

    plt.xlabel("Coin")
    plt.ylabel("Number of Trades")

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

coin_counts = get_coin_counts(
    final_eda_df
)

top_20_coins = get_top_coins(
    final_eda_df,
    top_n=20
)

coin_summary = calculate_coin_percentages(
    final_eda_df,
    top_n=20
)

top_3_concentration = (
    calculate_top_coin_concentration(
        final_eda_df,
        top_n=3
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 3 - STEP 7: COIN ANALYSIS")
print("=" * 70)

print("\nTOTAL UNIQUE COINS")
print("-" * 70)

print(
    final_eda_df["Coin"].nunique()
)

print("\n")
print("=" * 70)
print("TOP 20 MOST TRADED COINS")
print("=" * 70)

display(top_20_coins)

print("\n")
print("=" * 70)
print("TOP 20 COIN SUMMARY")
print("=" * 70)

display(coin_summary)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("COIN ANALYSIS INSIGHTS")
print("=" * 70)

top_coin = coin_summary.iloc[0]["Coin"]
top_coin_trades = coin_summary.iloc[0]["Trade Count"]
top_coin_percentage = coin_summary.iloc[0]["Percentage"]

print(
    f"Most Traded Coin: {top_coin}"
)

print(
    f"Trade Count: {top_coin_trades:,}"
)

print(
    f"Share of Total Trades: "
    f"{top_coin_percentage:.2f}%"
)

print(
    f"\nTop 3 Coins Account For "
    f"{top_3_concentration:.2f}% "
    f"of All Trades."
)

if top_3_concentration >= 50:
    print(
        "Observation: Trading activity is highly concentrated in a few assets."
    )

elif top_3_concentration >= 30:
    print(
        "Observation: Trading activity is moderately concentrated."
    )

else:
    print(
        "Observation: Trading activity is well distributed across assets."
    )

# ============================================================
# VISUALIZATION
# ============================================================

plot_top_coins(
    final_eda_df,
    top_n=20
)
```

TRADER ANALYSIS


```python
# ============================================================
# PHASE 3 - STEP 8: TRADER ANALYSIS
# Objective:
# Analyze trader participation and activity concentration.
#
# Feature:
# - Account
#
# Outputs:
# - Number of Unique Traders
# - Top 10 Active Traders
# - Trades-per-Trader Distribution
# - Trader Activity Statistics
#
# Insights:
# - Is trading activity concentrated among a few traders?
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def get_unique_trader_count(dataframe):
    """
    Count unique traders.
    """
    return dataframe["Account"].nunique()


def get_trades_per_trader(dataframe):
    """
    Calculate trade counts per trader.
    """
    trader_activity = (
        dataframe["Account"]
        .value_counts()
        .reset_index()
    )

    trader_activity.columns = [
        "Account",
        "Trade Count"
    ]

    return trader_activity


def get_top_active_traders(
    dataframe,
    top_n=10
):
    """
    Get most active traders.
    """
    top_traders = (
        dataframe["Account"]
        .value_counts()
        .head(top_n)
        .reset_index()
    )

    top_traders.columns = [
        "Account",
        "Trade Count"
    ]

    return top_traders


def get_trader_activity_statistics(
    trader_activity
):
    """
    Generate summary statistics for trader activity.
    """
    statistics = (
        trader_activity["Trade Count"]
        .describe()
    )

    return statistics


def calculate_top_trader_concentration(
    trader_activity,
    top_n=10
):
    """
    Calculate concentration of activity among top traders.
    """
    total_trades = (
        trader_activity["Trade Count"]
        .sum()
    )

    top_trades = (
        trader_activity["Trade Count"]
        .head(top_n)
        .sum()
    )

    concentration_percentage = (
        top_trades
        / total_trades
        * 100
    )

    return concentration_percentage


def plot_top_traders(
    trader_activity,
    top_n=10
):
    """
    Plot top active traders.
    """
    top_traders = trader_activity.head(top_n)

    plt.figure(figsize=(12, 6))

    plt.bar(
        range(len(top_traders)),
        top_traders["Trade Count"]
    )

    plt.title(
        f"Top {top_n} Most Active Traders"
    )

    plt.xlabel("Trader Rank")
    plt.ylabel("Number of Trades")

    plt.xticks(
        range(len(top_traders)),
        [f"T{i+1}" for i in range(len(top_traders))]
    )

    plt.tight_layout()
    plt.show()


def plot_trader_distribution(
    trader_activity
):
    """
    Plot distribution of trades per trader.
    """
    plt.figure(figsize=(10, 5))

    plt.hist(
        trader_activity["Trade Count"],
        bins=20
    )

    plt.title(
        "Trades per Trader Distribution"
    )

    plt.xlabel("Number of Trades")
    plt.ylabel("Number of Traders")

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

unique_traders = get_unique_trader_count(
    final_eda_df
)

trader_activity = get_trades_per_trader(
    final_eda_df
)

top_10_traders = get_top_active_traders(
    final_eda_df,
    top_n=10
)

trader_statistics = (
    get_trader_activity_statistics(
        trader_activity
    )
)

top_10_concentration = (
    calculate_top_trader_concentration(
        trader_activity,
        top_n=10
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 3 - STEP 8: TRADER ANALYSIS")
print("=" * 70)

print(
    f"\nNumber of Unique Traders: "
    f"{unique_traders}"
)

print("\n")
print("=" * 70)
print("TOP 10 ACTIVE TRADERS")
print("=" * 70)

display(top_10_traders)

print("\n")
print("=" * 70)
print("TRADER ACTIVITY STATISTICS")
print("=" * 70)

print(trader_statistics)

print("\n")
print("=" * 70)
print("TRADES PER TRADER")
print("=" * 70)

display(
    trader_activity.head(20)
)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("TRADER ANALYSIS INSIGHTS")
print("=" * 70)

print(
    f"Total Unique Traders: "
    f"{unique_traders}"
)

print(
    f"Top 10 Traders Account For "
    f"{top_10_concentration:.2f}% "
    f"of All Trades."
)

most_active_trader = (
    trader_activity.iloc[0]["Trade Count"]
)

least_active_trader = (
    trader_activity.iloc[-1]["Trade Count"]
)

print(
    f"Most Active Trader Trades : "
    f"{most_active_trader:,}"
)

print(
    f"Least Active Trader Trades: "
    f"{least_active_trader:,}"
)

if top_10_concentration >= 70:
    print(
        "Observation: Trading activity is highly concentrated among a small group of traders."
    )

elif top_10_concentration >= 50:
    print(
        "Observation: Trading activity is moderately concentrated among top traders."
    )

else:
    print(
        "Observation: Trading activity is relatively well distributed across traders."
    )

# ============================================================
# VISUALIZATIONS
# ============================================================

plot_top_traders(
    trader_activity,
    top_n=10
)

plot_trader_distribution(
    trader_activity
)
```

TIME-BASED ANALYSIS


```python
# ============================================================
# PHASE 3 - STEP 9: TIME-BASED ANALYSIS
# Objective:
# Analyze trading activity over time.
#
# Feature:
# - Merge_Date
#
# Outputs:
# - Daily Trade Counts
# - Monthly Trade Counts
# - Daily Activity Plot
# - Monthly Activity Plot
#
# Insights:
# - Activity Spikes
# - Market Participation Trends
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt


def preprocess_merge_date(dataframe):
    """
    Convert Merge_Date to datetime.
    """
    dataframe = dataframe.copy()

    dataframe["Merge_Date"] = pd.to_datetime(
        dataframe["Merge_Date"]
    )

    return dataframe


def get_daily_trade_counts(dataframe):
    """
    Calculate daily trading activity.
    """
    daily_trades = (
        dataframe
        .groupby("Merge_Date")
        .size()
        .reset_index(name="Trade Count")
        .sort_values("Merge_Date")
    )

    return daily_trades


def get_monthly_trade_counts(dataframe):
    """
    Calculate monthly trading activity.
    """
    monthly_trades = (
        dataframe
        .groupby(
            dataframe["Merge_Date"].dt.to_period("M")
        )
        .size()
        .reset_index(name="Trade Count")
    )

    monthly_trades["Merge_Date"] = (
        monthly_trades["Merge_Date"]
        .astype(str)
    )

    return monthly_trades


def get_daily_activity_statistics(
    daily_trades
):
    """
    Generate daily activity statistics.
    """
    return daily_trades["Trade Count"].describe()


def identify_activity_spikes(
    daily_trades,
    top_n=10
):
    """
    Identify highest activity days.
    """
    activity_spikes = (
        daily_trades
        .sort_values(
            by="Trade Count",
            ascending=False
        )
        .head(top_n)
    )

    return activity_spikes


def plot_daily_trade_activity(
    daily_trades
):
    """
    Plot daily trading activity.
    """
    plt.figure(figsize=(14, 6))

    plt.plot(
        daily_trades["Merge_Date"],
        daily_trades["Trade Count"]
    )

    plt.title(
        "Daily Trading Activity"
    )

    plt.xlabel("Date")
    plt.ylabel("Number of Trades")

    plt.tight_layout()
    plt.show()


def plot_monthly_trade_activity(
    monthly_trades
):
    """
    Plot monthly trading activity.
    """
    plt.figure(figsize=(14, 6))

    plt.bar(
        monthly_trades["Merge_Date"],
        monthly_trades["Trade Count"]
    )

    plt.title(
        "Monthly Trading Activity"
    )

    plt.xlabel("Month")
    plt.ylabel("Number of Trades")

    plt.xticks(rotation=90)

    plt.tight_layout()
    plt.show()


# ============================================================
# MAIN PROGRAM
# ============================================================

final_eda_df = preprocess_merge_date(
    final_eda_df
)

daily_trades = get_daily_trade_counts(
    final_eda_df
)

monthly_trades = get_monthly_trade_counts(
    final_eda_df
)

daily_statistics = (
    get_daily_activity_statistics(
        daily_trades
    )
)

activity_spikes = (
    identify_activity_spikes(
        daily_trades,
        top_n=10
    )
)

# ============================================================
# OUTPUTS
# ============================================================

print("=" * 70)
print("PHASE 3 - STEP 9: TIME-BASED ANALYSIS")
print("=" * 70)

print("\n")
print("=" * 70)
print("DAILY TRADE ACTIVITY SUMMARY")
print("=" * 70)

print(daily_statistics)

print("\n")
print("=" * 70)
print("TOP 10 ACTIVITY SPIKES")
print("=" * 70)

display(activity_spikes)

print("\n")
print("=" * 70)
print("MONTHLY TRADE COUNTS")
print("=" * 70)

display(monthly_trades)

# ============================================================
# INSIGHTS
# ============================================================

print("\n" + "=" * 70)
print("TIME-BASED ANALYSIS INSIGHTS")
print("=" * 70)

highest_activity_day = (
    activity_spikes.iloc[0]["Merge_Date"]
)

highest_activity_count = (
    activity_spikes.iloc[0]["Trade Count"]
)

lowest_activity_day = (
    daily_trades.loc[
        daily_trades["Trade Count"].idxmin()
    ]
)

print(
    f"Highest Activity Day: "
    f"{highest_activity_day.date()}"
)

print(
    f"Trades on Highest Activity Day: "
    f"{highest_activity_count:,}"
)

print(
    f"\nLowest Activity Day: "
    f"{lowest_activity_day['Merge_Date'].date()}"
)

print(
    f"Trades on Lowest Activity Day: "
    f"{lowest_activity_day['Trade Count']:,}"
)

monthly_growth = (
    monthly_trades["Trade Count"]
    .iloc[-1]
    -
    monthly_trades["Trade Count"]
    .iloc[0]
)

if monthly_growth > 0:
    print(
        "\nObservation: Trading participation generally increased over time."
    )

elif monthly_growth < 0:
    print(
        "\nObservation: Trading participation generally decreased over time."
    )

else:
    print(
        "\nObservation: Trading participation remained relatively stable."
    )

print(
    "\nObservation: Activity spikes may correspond to periods of "
    "high market volatility, major crypto events, or strong sentiment shifts."
)

# ============================================================
# VISUALIZATIONS
# ============================================================

plot_daily_trade_activity(
    daily_trades
)

plot_monthly_trade_activity(
    monthly_trades
)
```
